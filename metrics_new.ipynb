```python
import os
import json
import pandas as pd

# --- Configuration ---
root_dir = "experiments/flux_schnell/style/generated_images/" # Update to your base path # If your paths start with a shared prefix
concepts = ['cyberpunk', 'anime', 'sketch', 'impressionism'] # Add your actual concepts

facts = ['300_1_mean_diff_s_only_pooled', 
         '300_1_mean_diff_s_only_pooled_0_7',
         '300_1_mean_diff_s_only_pooled_0_9',
         '300_1_mean_diff_s_only_pooled_0_12'] # Add your fact identifiers

data_list = []

# --- Main Logic ---
for concept in concepts:
    for fact in facts:
        # Construct the directory name based on your pattern
        # Adjust the underscores "_" below if your naming convention differs
        
        dir_path = os.path.join(root_dir, concept, fact)
        file_path = os.path.join(dir_path, "summary_metrics.json")
        
        if os.path.exists(file_path):
            try:
                with open(file_path, 'r') as f:
                    metrics = json.load(f)
                
                # Add concept and fact info to the metric dict for tracking
                metrics['concept'] = concept
                metrics['blocks'] = fact
                
                data_list.append(metrics)
                print(f"✅ Loaded: {concept}, {fact}")
            except Exception as e:
                print(f"❌ Error reading {file_path}: {e}")
        else:
            print(f"⚠️ Missing: {dir_path}")

# --- Create Table ---
df = pd.DataFrame(data_list)

# Reorder columns to put Concept and Fact at the front
cols = ['concept', 'blocks'] + [c for c in df.columns if c not in ['concept', 'blocks']]
df = df[cols]

# Save for plotting
df.to_csv("all_metrics_combined.csv", index=False)
print("\n--- Final Data Table ---")
print(df)
```
