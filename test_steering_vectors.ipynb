```python
import torch
from tabulate import tabulate # Use 'pip install tabulate' for clean tables
import pandas as pd
from tqdm import tqdm

step_tables = []
for i, key in enumerate(pos_concept):

    table_data = []
    print(f"\n### STEP {i + 1} EIGENVALUES ###")
    results = []
    for layer_idx, layer in tqdm(enumerate(pos_concept[i])):
        pos = pos_concept[i][f'{layer}'].to('cuda')
        neg = neg_concept[i][f'{layer}'].to('cuda')

        diff = (pos - neg).mean(dim=(1,2))
        
        diff = diff.float()
        # Center the data for true PCA singular values
        layer_centered = diff - torch.mean(diff, dim=0)
        
        # svdvals is faster than full svd
        # It returns singular values in descending order
        eigenvalues = torch.linalg.svdvals(layer_centered)
        
        # Grab the top 3
        top_3 = eigenvalues[:3].tolist()
        
        results.append({
            "Layer": f"Layer {layer_idx + 1}",
            "Eigen_1": top_3[0],
            "Eigen_2": top_3[1],
            "Eigen_3": top_3[2]
        })
    
    # Create DataFrame for the current step
    df = pd.DataFrame(results).set_index("Layer")
    step_tables.append(df)
    # Display the table for the current step
```

```python
import torch.nn.functional as F

def get_coherence_score(act_target, act_control):
    # 1. Calculate individual directions
    directions = act_target - act_control # [batch, seq, d_model]
    # 2. Calculate the aggregate mean direction
    mean_dir = directions.mean(dim=0, keepdim=True) 
    # 3. Calculate cosine similarity of each pair to the mean
    cos_sim = F.cosine_similarity(directions, mean_dir, dim=-1)
    return cos_sim.mean().item() # High score = stable direction
```

```python
import torch.nn.functional as F

def get_coherence_score(act_target, act_control, norm=False):
    
    # 1. Calculate individual directions
    directions = act_target - act_control # [batch, seq, d_model]
    
    if norm:
        directions = directions / directions.norm(dim=-1, keepdim=True)
    # 2. Calculate the aggregate mean direction
    mean_dir = directions.mean(dim=0, keepdim=True) 

    if norm:
        directions = directions / directions.norm(dim=-1, keepdim=True)
        mean_dir = mean_dir / mean_dir.norm(dim=-1, keepdim=True)
    # 3. Calculate cosine similarity of each pair to the mean
    cos_sim = F.cosine_similarity(directions, mean_dir, dim=-1)
    return cos_sim.mean().item() # High score = stable direction
```
