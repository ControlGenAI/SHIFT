## step 0 layers all

```python
import math
# ===== config =====
STEP = 3
BRANCH = "txt"      # "txt" or "img"
N_SAMPLES = 20
D_MAX = 64
CENTER = False      # True => SVD(delta - mu_delta), False => SVD(delta)
NCOLS = 5
pos_data = torch.load(POS_ACT_PATH, map_location="cpu")
neg_data = torch.load(NEG_ACT_PATH, map_location="cpu")
layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if k.startswith("layer_")],
    key=lambda x: int(x.split("_")[1])
)
n_layers = len(layer_keys)
nrows = math.ceil(n_layers / NCOLS)
fig, axes = plt.subplots(nrows, NCOLS, figsize=(NCOLS * 3.4, nrows * 2.8), squeeze=False)
for i, layer_key in enumerate(layer_keys):
    r, c = divmod(i, NCOLS)
    ax = axes[r][c]
    x_pos = pos_data[STEP][layer_key]
    x_neg = neg_data[STEP][layer_key]
    
    if isinstance(x_pos, dict):   # dual branch format
        x_pos = x_pos[BRANCH].mean(dim=0)
        x_neg = x_neg[BRANCH].mean(dim=0)
    
    print(x_pos.shape, x_neg.shape)
    if x_pos.dim() == 4 and x_pos.shape[1] == 1:
        x_pos = x_pos.squeeze(1)
    if x_neg.dim() == 4 and x_neg.shape[1] == 1:
        x_neg = x_neg.squeeze(1)
    n = min(N_SAMPLES, x_pos.shape[0], x_neg.shape[0])
    pos = x_pos[:n].float()
    neg = x_neg[:n].float()
    delta = (pos - neg).reshape(-1, pos.shape[-1])  # (N*T, C)
    mu_delta = delta.mean(dim=0)
    X = delta - mu_delta.unsqueeze(0) if CENTER else delta
    _, S, Vh = torch.linalg.svd(X, full_matrices=False)
    evr = (S.pow(2).cumsum(0) / S.pow(2).sum().clamp(min=1e-12))[:D_MAX]
    d90 = int((evr < 0.90).sum().item()) + 1
    pc1 = Vh[0]
    cos = torch.nn.functional.cosine_similarity(mu_delta.unsqueeze(0), pc1.unsqueeze(0)).item()
    ax.plot(torch.arange(1, len(evr) + 1), evr.numpy(), linewidth=1.5)
    ax.axhline(0.9, linestyle="--", linewidth=1)
    ax.set_ylim(0, 1.02)
    ax.set_title(f"{layer_key}\nd90={d90}, cos={cos:.2f}", fontsize=8)
    ax.grid(alpha=0.25)
# hide empty axes
for j in range(n_layers, nrows * NCOLS):
    r, c = divmod(j, NCOLS)
    axes[r][c].axis("off")
fig.suptitle(f"EVR per layer | step={STEP} | branch={BRANCH} | center={CENTER}", fontsize=12)
plt.tight_layout()
plt.show()
```

## Реконструкционная ошибка vs d

```python
# ==========================================================
# Check 1: Reconstruction error vs d
# ==========================================================
# Expects:
#   pos_acts, neg_acts: torch.Tensor of shape (N, T, C)
# Example:
#   pos_acts = pos_data[step][layer]["txt"].squeeze(1).float()[:N]
#   neg_acts = neg_data[step][layer]["txt"].squeeze(1).float()[:N]

import torch
import matplotlib.pyplot as plt

def reconstruction_error_vs_d(
    pos_acts: torch.Tensor,
    neg_acts: torch.Tensor,
    d_max: int = 50,
    center: bool = False,
):
    """
    Returns:
      ds            : list[int]
      rel_errors    : list[float]
      svals         : torch.Tensor (all singular values)
      delta_ref     : (T, C) mean shift used as target
      delta_centered: matrix actually decomposed by SVD
    """
    assert pos_acts.dim() == 3 and neg_acts.dim() == 3, (pos_acts.shape, neg_acts.shape)
    assert pos_acts.shape == neg_acts.shape, (pos_acts.shape, neg_acts.shape)

    # Mean paired token-wise shift (T, C)
    delta_ref = (pos_acts - neg_acts).mean(dim=0).float()  # (T, C)

    # Matrix used in SVD
    if center:
        row_mean = delta_ref.mean(dim=0, keepdim=True)     # (1, C)
        X = delta_ref - row_mean
    else:
        X = delta_ref

    # SVD: X = U @ diag(S) @ Vh
    U, S, Vh = torch.linalg.svd(X, full_matrices=False)
    r = S.numel()
    d_max_eff = min(int(d_max), int(r))

    denom = delta_ref.norm().clamp(min=1e-12)
    ds, rel_errors = [], []

    for d in range(1, d_max_eff + 1):
        # Rank-d reconstruction of X
        X_approx = (U[:, :d] * S[:d]) @ Vh[:d, :]          # (T, C)

        # Compare against original delta_ref (undo centering if needed)
        if center:
            delta_approx = X_approx + row_mean
        else:
            delta_approx = X_approx

        err = (delta_ref - delta_approx).norm() / denom
        ds.append(d)
        rel_errors.append(float(err))

    return ds, rel_errors, S, delta_ref, X


# ===== Example usage =====
# ds, rel_errors, S, delta_ref, X = reconstruction_error_vs_d(
#     pos_acts, neg_acts, d_max=50, center=False
# )
# plt.figure(figsize=(6,4))
# plt.plot(ds, rel_errors, marker="o", markersize=3)
# plt.xlabel("d")
# plt.ylabel("relative reconstruction error")
# plt.title("Reconstruction error vs d")
# plt.grid(alpha=0.3)
# plt.tight_layout()
# plt.show()
```

```python

def _to_ntc(x):
    if x.dim() == 4 and x.shape[1] == 1:
        x = x.squeeze(1)  # (N, T, C)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,T,C), got {tuple(x.shape)}")
    return x.float()
def _get_branch_tensor(data, step, layer_key, branch="txt"):
    x = data[step][layer_key]
    if isinstance(x, dict):
        x = x[branch]
    return _to_ntc(x)
def layer_metrics(pos_ntc, neg_ntc, d_max=64, center=False):
    # pos_ntc, neg_ntc: (N,T,C)
    n = min(pos_ntc.shape[0], neg_ntc.shape[0], N_SAMPLES)
    pos_ntc = pos_ntc[:n]
    neg_ntc = neg_ntc[:n]
    # Mean paired token-wise shift (T,C)
    delta_ref = (pos_ntc - neg_ntc).mean(dim=0).float()
    if center:
        row_mean = delta_ref.mean(dim=0, keepdim=True)     # (1,C)
        X = delta_ref - row_mean
    else:
        row_mean = None
        X = delta_ref
    # SVD: X = U @ diag(S) @ Vh
    U, S, Vh = torch.linalg.svd(X, full_matrices=False)
    r = int(S.numel())
    d_max_eff = min(int(d_max), r)
    ev = S.pow(2)
    cum_evr = ev.cumsum(0) / ev.sum().clamp(min=1e-12)
    cum_evr = cum_evr[:d_max_eff]
    # cos(mu_delta, PC1)
    mu_delta = delta_ref.mean(dim=0)  # mean over token rows => (C,)
    pc1 = Vh[0]
    cos = torch.nn.functional.cosine_similarity(mu_delta.unsqueeze(0), pc1.unsqueeze(0)).item()
    # reconstruction error vs d (against delta_ref)
    denom = delta_ref.norm().clamp(min=1e-12)
    rel_err = []
    ds = list(range(1, d_max_eff + 1))
    for d in ds:
        X_approx = (U[:, :d] * S[:d]) @ Vh[:d, :]
        if center:
            delta_approx = X_approx + row_mean
        else:
            delta_approx = X_approx
        err = (delta_ref - delta_approx).norm() / denom
        rel_err.append(float(err))
    d90 = int((cum_evr < 0.90).sum().item()) + 1
    return ds, rel_err, cum_evr.cpu(), d90, cos
# ---------- Load ----------
pos_data = torch.load(POS_ACT_PATH, map_location="cpu")
neg_data = torch.load(NEG_ACT_PATH, map_location="cpu")
layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if k.startswith("layer_")],
    key=lambda s: int(s.split("_")[1])
)
# ---------- Plot Grid ----------
n_layers = len(layer_keys)
nrows = math.ceil(n_layers / NCOLS)
fig, axes = plt.subplots(nrows, NCOLS, figsize=(NCOLS * 4.0, nrows * 3.1), squeeze=False)
for idx, layer_key in enumerate(layer_keys):
    r, c = divmod(idx, NCOLS)
    ax = axes[r][c]
    try:
        pos_ntc = _get_branch_tensor(pos_data, STEP, layer_key, branch=BRANCH)
        neg_ntc = _get_branch_tensor(neg_data, STEP, layer_key, branch=BRANCH)
        ds, rel_err, cum_evr, d90, cos = layer_metrics(pos_ntc, neg_ntc, d_max=D_MAX, center=CENTER)
        # left axis: reconstruction error
        ax.plot(ds, rel_err, color="tab:red", linewidth=1.7, label="rel err")
        ax.set_ylim(0.0, 1.05)
        ax.set_xlabel("d", fontsize=8)
        ax.set_ylabel("err", fontsize=8, color="tab:red")
        ax.tick_params(axis="y", labelcolor="tab:red", labelsize=7)
        ax.tick_params(axis="x", labelsize=7)
        ax.grid(alpha=0.25)
        # right axis: EVR
        ax2 = ax.twinx()
        ax2.plot(ds, cum_evr.numpy(), color="tab:blue", linewidth=1.5, label="cum EVR")
        ax2.axhline(0.9, linestyle="--", linewidth=1)
        ax2.set_ylim(0.0, 1.02)
        ax2.set_ylabel("EVR", fontsize=8, color="tab:blue")
        ax2.tick_params(axis="y", labelcolor="tab:blue", labelsize=7)
        ax.set_title(f"{layer_key}\nd90={d90}, cos={cos:.2f}", fontsize=8)
    except Exception as e:
        ax.axis("off")
        ax.set_title(f"{layer_key}\nERROR: {str(e)[:40]}...", fontsize=8)
# hide empty cells
for j in range(n_layers, nrows * NCOLS):
    rr, cc = divmod(j, NCOLS)
    axes[rr][cc].axis("off")
fig.suptitle(
    f"Step={STEP} | branch={BRANCH} | center={CENTER}\n"
    f"red: reconstruction error, blue: cumulative EVR",
    fontsize=12
)
plt.tight_layout()
plt.show()
```

## Устойчивость по промптам

```python

def _to_ntc(x):
    # (N,1,T,C) -> (N,T,C)
    if x.dim() == 4 and x.shape[1] == 1:
        x = x.squeeze(1)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,T,C), got {tuple(x.shape)}")
    return x.float()
def _extract_layer_branch(data, step, layer_key, branch="txt"):
    x = data[step][layer_key]
    if isinstance(x, dict):
        x = x[branch]
    return _to_ntc(x)
def top_dir_cosine_matrix(pos_ntc, neg_ntc, n_samples=20, use_abs=True, center_each_pair=False):
    """
    pos_ntc, neg_ntc: (N,T,C)
    Returns:
      cos_mat: (n,n)
      pair_vals: flattened upper-tri values
    """
    n = min(n_samples, pos_ntc.shape[0], neg_ntc.shape[0])
    pos_ntc = pos_ntc[:n]
    neg_ntc = neg_ntc[:n]
    top_dirs = []
    for k in range(n):
        delta_k = (pos_ntc[k] - neg_ntc[k]).float()   # (T,C)
        if center_each_pair:
            delta_k = delta_k - delta_k.mean(dim=0, keepdim=True)
        # SVD on (T,C), right singular vectors in feature space C
        _, _, Vh_k = torch.linalg.svd(delta_k, full_matrices=False)
        v = Vh_k[0]
        v = v / v.norm().clamp(min=1e-12)
        top_dirs.append(v)
    top_dirs = torch.stack(top_dirs, dim=0)            # (n,C)
    cos_mat = top_dirs @ top_dirs.T                    # normalized => cosine matrix
    if use_abs:
        cos_mat = cos_mat.abs()
    vals = []
    for i in range(n):
        for j in range(i + 1, n):
            vals.append(float(cos_mat[i, j]))
    vals = torch.tensor(vals) if len(vals) else torch.tensor([])
    return cos_mat.cpu(), vals
# ---------- load ----------
USE_ABS_COS = True      # обычно лучше True, т.к. знак сингулярного вектора произволен
CENTER_EACH_PAIR = False

pos_data = torch.load(POS_ACT_PATH, map_location="cpu")
neg_data = torch.load(NEG_ACT_PATH, map_location="cpu")
layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if k.startswith("layer_")],
    key=lambda s: int(s.split("_")[1])
)
# ---------- compute for all layers ----------
cos_by_layer = {}
stats = []  # (layer_idx, median, q25, q75, mean, std)
for layer_key in layer_keys:
    pos_ntc = _extract_layer_branch(pos_data, STEP, layer_key, branch=BRANCH)
    neg_ntc = _extract_layer_branch(neg_data, STEP, layer_key, branch=BRANCH)
    cos_mat, vals = top_dir_cosine_matrix(
        pos_ntc, neg_ntc,
        n_samples=N_SAMPLES,
        use_abs=USE_ABS_COS,
        center_each_pair=CENTER_EACH_PAIR,
    )
    cos_by_layer[layer_key] = cos_mat
    if vals.numel() > 0:
        q25 = float(torch.quantile(vals, 0.25))
        med = float(torch.quantile(vals, 0.50))
        q75 = float(torch.quantile(vals, 0.75))
        mean = float(vals.mean())
        std = float(vals.std(unbiased=False))
    else:
        q25 = med = q75 = mean = std = float("nan")
    layer_idx = int(layer_key.split("_")[1])
    stats.append((layer_idx, med, q25, q75, mean, std))
stats = sorted(stats, key=lambda x: x[0])
# =======================
# 1) Grid heatmap per layer
# =======================
n_layers = len(layer_keys)
nrows = math.ceil(n_layers / NCOLS)
fig, axes = plt.subplots(nrows, NCOLS, figsize=(NCOLS * 3.2, nrows * 3.0), squeeze=False)
for i, layer_key in enumerate(layer_keys):
    r, c = divmod(i, NCOLS)
    ax = axes[r][c]
    cm = cos_by_layer[layer_key]
    im = ax.imshow(cm.numpy(), vmin=0.0 if USE_ABS_COS else -1.0, vmax=1.0, cmap="viridis")
    idx = int(layer_key.split("_")[1])
    # quick stat in title
    vals = cm[torch.triu(torch.ones_like(cm), diagonal=1) > 0]
    med = float(torch.median(vals)) if vals.numel() else float("nan")
    ax.set_title(f"layer_{idx}\nmed={'{:.2f}'.format(med)}", fontsize=8)
    ax.set_xticks([])
    ax.set_yticks([])
for j in range(n_layers, nrows * NCOLS):
    rr, cc = divmod(j, NCOLS)
    axes[rr][cc].axis("off")
cbar = fig.colorbar(im, ax=axes, fraction=0.015, pad=0.02)
cbar.set_label("|cos|" if USE_ABS_COS else "cos", rotation=90)
fig.suptitle(
    f"Top-1 direction stability across prompts | step={STEP}, branch={BRANCH}, abs={USE_ABS_COS}",
    fontsize=12
)
plt.tight_layout()
plt.show()

# =======================
# 2) Summary across layers (median + IQR)
# =======================
layer_ids = [x[0] for x in stats]
medians   = [x[1] for x in stats]
q25s      = [x[2] for x in stats]
q75s      = [x[3] for x in stats]
means     = [x[4] for x in stats]
# asymmetric error bars for IQR
yerr_lower = [m - q for m, q in zip(medians, q25s)]
yerr_upper = [q - m for m, q in zip(medians, q75s)]
plt.figure(figsize=(10, 4))
plt.errorbar(layer_ids, medians, yerr=[yerr_lower, yerr_upper], fmt='o-', capsize=3, label='median ± IQR')
plt.plot(layer_ids, means, '--', alpha=0.7, label='mean')
plt.ylim(0.0 if USE_ABS_COS else -1.0, 1.0)
plt.xlabel("Layer")
plt.ylabel("|cos| pairwise top-dir" if USE_ABS_COS else "cos pairwise top-dir")
plt.title(f"Direction stability by layer | step={STEP}, branch={BRANCH}")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import torch
import matplotlib.pyplot as plt

# neg_acts, pos_acts: (N, T, C) — одна пара архивов, один слой, один step
# N = число промптов (пар)

def pooled_activation_delta(pos_acts, neg_acts):
    """Одна «стрелка» на промпт в полном C: mean_t (a+ - a-)."""
    dlt = (pos_acts.float() - neg_acts.float()).mean(dim=1)  # (N, C)
    return dlt / dlt.norm(dim=1, keepdim=True).clamp(min=1e-8)


def pairwise_cosine_matrix(X):
    """X: (N, D) L2-нормированные строки → косинусы = X @ X.T"""
    return (X @ X.T).cpu()


def plot_cosine_hist(G):
    N = G.shape[0]
    iu = torch.triu_indices(N, N, offset=1)
    vals = G[iu[0], iu[1]].numpy()
    plt.figure(figsize=(6, 3))
    plt.hist(vals, bins=40, color="steelblue", alpha=0.85)
    plt.xlabel("cos(Δ_i, Δ_j)")
    plt.ylabel("count")
    plt.title("Pairwise cosine between per-prompt steering directions")
    plt.tight_layout()
    plt.show()
    print(f"pairwise cos: mean={vals.mean():.3f}, std={vals.std():.3f}, "
          f"min={vals.min():.3f}, max={vals.max():.3f}")


# --- 1) Вектор на промпт ---
Delta = pooled_activation_delta(pos_acts, neg_acts)  # (N, C), нормированы

G = pairwise_cosine_matrix(Delta)
plot_cosine_hist(G)

# --- 2) Кластеризация в R^C (или сначала PCA для скорости) ---
try:
    from sklearn.cluster import KMeans
    from sklearn.decomposition import PCA
    from sklearn.metrics import silhouette_score
except ImportError:
    raise RuntimeError("pip install scikit-learn")

X = Delta.cpu().numpy()
# опционально: сжать перед k-means (быстрее и стабильнее при большом C)
pca_pre = PCA(n_components=min(64, X.shape[1], X.shape[0] - 1), random_state=0)
Xz = pca_pre.fit_transform(X)

for K in [2, 3, 4]:
    if Xz.shape[0] < K + 1:
        continue
    km = KMeans(n_clusters=K, n_init=20, random_state=0)
    lab = km.fit_predict(Xz)
    sil = silhouette_score(Xz, lab)
    print(f"K={K}  silhouette={sil:.3f}  counts={[(lab == j).sum() for j in range(K)]}")

# визуализация кластеров в 2D (PCA от Delta)
pca2 = PCA(n_components=2, random_state=0)
XY = pca2.fit_transform(X)
K_plot = 3
km = KMeans(n_clusters=K_plot, n_init=20, random_state=0)
lab = km.fit_predict(Xz)
plt.figure(figsize=(5, 4))
for j in range(K_plot):
    m = lab == j
    plt.scatter(XY[m, 0], XY[m, 1], s=18, alpha=0.75, label=f"cluster {j}")
plt.legend(markerscale=1.2, fontsize=8)
plt.title("Per-prompt Δ (pooled) — PCA-2D, colors = KMeans")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.tight_layout()
plt.show()
```

```python
import torch

# pos_acts, neg_acts: (N, T, C) для выбранного step/layer/branch
# Пример получения:
# pos_acts = ...
# neg_acts = ...

def fit_U_from_acts(
    pos_acts: torch.Tensor,
    neg_acts: torch.Tensor,
    d: int = 32,
    pca_delta_mode: str = "raw_delta",   # "raw_delta" | "centered" | "append_mean_dir"
):
    """
    Возвращает:
      U      : (C, d_eff)
      mu_neg : (T, C)   (mean over batch, tokenwise)
      S      : сингулярные значения
      Vh     : правые сингулярные векторы (d_full, C)
    """
    assert pos_acts.shape == neg_acts.shape and pos_acts.dim() == 3
    N, T, C = pos_acts.shape

    pos = pos_acts.float()
    neg = neg_acts.float()

    # tokenwise anchor (как ты и хотела: только mean over batch)
    mu_neg = neg.mean(dim=0)                         # (T, C)

    # paired deltas
    delta = pos - neg                                # (N, T, C)
    X = delta.reshape(N * T, C)                      # (N*T, C)

    if pca_delta_mode == "centered":
        X = X - X.mean(dim=0, keepdim=True)
    elif pca_delta_mode == "append_mean_dir":
        m = X.mean(dim=0, keepdim=True)              # (1, C)
        Xc = X - m
        # усиливаем mean direction, чтобы точно попасть в подпространство
        X = torch.cat([Xc, m], dim=0)
    elif pca_delta_mode == "raw_delta":
        pass
    else:
        raise ValueError(f"Unknown pca_delta_mode={pca_delta_mode}")

    # SVD по фичам C
    # X = U_s @ diag(S) @ Vh, где строки Vh — базисные направления в R^C
    _, S, Vh = torch.linalg.svd(X, full_matrices=False)

    d_eff = min(int(d), int(Vh.shape[0]))
    U = Vh[:d_eff].T.contiguous()                    # (C, d_eff)

    return U, mu_neg, S, Vh


# ===== usage =====
U, mu_neg, S, Vh = fit_U_from_acts(
    pos_acts, neg_acts,
    d=32,
    pca_delta_mode="raw_delta",
)
print("U:", U.shape, "mu_neg:", mu_neg.shape)
```

```python
import torch
from src.steering.subspace_act import ContrastivePCATokenWise, AffineTransport


def run_split_eval(Z_neg, Z_pos, train_frac=0.8, seed=42):
    """Z_neg, Z_pos: (N, T, d) — один random split."""
    g = torch.Generator(device=Z_neg.device)
    g.manual_seed(seed)

    N, T, d = Z_neg.shape
    n_train = int(N * train_frac)
    idx = torch.randperm(N, generator=g, device=Z_neg.device)
    train_idx = idx[:n_train]
    val_idx = idx[n_train:]

    Ws_train, bs_train = [], []
    for t in range(T):
        tr = AffineTransport().fit(
            Z_neg[train_idx, t, :],
            Z_pos[train_idx, t, :],
        )
        Ws_train.append(tr.W)
        bs_train.append(tr.b)
    Ws_train = torch.stack(Ws_train)  # (T, d, d)
    bs_train = torch.stack(bs_train)  # (T, d)

    Z_neg_val = Z_neg[val_idx]
    Z_pos_val = Z_pos[val_idx]

    Z_pred = torch.einsum("ntd,tdk->ntk", Z_neg_val, Ws_train) + bs_train.unsqueeze(0)
    mean_diff = (Z_pos[train_idx] - Z_neg[train_idx]).mean(dim=0)
    Z_pred_baseline = Z_neg_val + mean_diff.unsqueeze(0)

    err_transport_per_sample = (Z_pred - Z_pos_val).norm(dim=-1).mean(dim=-1)
    err_baseline_per_sample = (Z_pred_baseline - Z_pos_val).norm(dim=-1).mean(dim=-1)

    return {
        "err_transport": err_transport_per_sample.mean().item(),
        "err_baseline": err_baseline_per_sample.mean().item(),
        "improve_pct": (
            (err_baseline_per_sample.mean() - err_transport_per_sample.mean())
            / (err_baseline_per_sample.mean() + 1e-12)
            * 100
        ).item(),
    }


def eval_layer_repeated(
    neg_acts,
    pos_acts,
    *,
    subspace_dim=5,
    pca_delta_mode="centered",
    train_frac=0.8,
    num_splits=20,
    seed0=1000,
    min_n=4,
):
    """
    neg_acts, pos_acts: (N, T, C) на одном слое.
    CPCA fit на всех N (как при сохранении вектора); split только для affine / baseline.
    """
    n = min(neg_acts.shape[0], pos_acts.shape[0])
    if n < min_n:
        return None
    neg_acts = neg_acts[:n].float()
    pos_acts = pos_acts[:n].float()

    cpca = ContrastivePCATokenWise(
        subspace_dim=subspace_dim,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_acts, pos_acts)

    Z_neg = cpca.project(neg_acts)
    Z_pos = cpca.project(pos_acts)

    outs = [
        run_split_eval(Z_neg, Z_pos, train_frac=train_frac, seed=seed0 + s)
        for s in range(num_splits)
    ]
    te = torch.tensor([o["err_transport"] for o in outs])
    be = torch.tensor([o["err_baseline"] for o in outs])
    im = torch.tensor([o["improve_pct"] for o in outs])
    return {
        "d": int(cpca.U.shape[1]),
        "T": int(Z_neg.shape[1]),
        "N": int(Z_neg.shape[0]),
        "transport_mean": te.mean().item(),
        "transport_std": te.std(unbiased=True).item(),
        "baseline_mean": be.mean().item(),
        "baseline_std": be.std(unbiased=True).item(),
        "improve_mean_pct": im.mean().item(),
        "improve_std_pct": im.std(unbiased=True).item(),
        "transport_wins_pct": (te < be).float().mean().item() * 100.0,
    }


def eval_all_layers_one_step(
    pos,
    neg,
    step,
    *,
    blocks=19,
    subspace_dim=32,
    pca_delta_mode="centered",
    num_splits=20,
    train_frac=0.8,
):
    rows = []
    for block in range(blocks):
        layer = f"layer_{block}"
        if step not in pos or step not in neg:
            continue
        if layer not in pos[step] or layer not in neg[step]:
            continue
        pn = pos[step][layer]
        nn = neg[step][layer]
        # при dual: pos[step][layer]['txt'] и т.д.
        if isinstance(pn, dict):
            if "txt" in pn:
                pn, nn = pn["txt"], nn["txt"]
            elif "img" in pn:
                pn, nn = pn["img"], nn["img"]
            else:
                continue
        if not (torch.is_tensor(pn) and torch.is_tensor(nn)):
            continue
        while pn.dim() > 3:
            pn = pn.squeeze(1)
            nn = nn.squeeze(1)
        if pn.dim() != 3 or nn.dim() != 3:
            continue

        stats = eval_layer_repeated(
            nn,
            pn,
            subspace_dim=subspace_dim,
            pca_delta_mode=pca_delta_mode,
            train_frac=train_frac,
            num_splits=num_splits,
        )
        if stats is None:
            continue
        rows.append((layer, stats))

    # печать таблицей
    print(f"step={step}  subspace_dim={subspace_dim}  pca_delta_mode={pca_delta_mode}  num_splits={num_splits}\n")
    hdr = f"{'layer':<12} {'N':>4} {'T':>4} {'d':>3} | {'tr_err':>8} {'±':>5} | {'base':>8} {'±':>5} | {'imp%':>7} {'win%':>6}"
    print(hdr)
    print("-" * len(hdr))
    for layer, s in rows:
        print(
            f"{layer:<12} {s['N']:>4} {s['T']:>4} {s['d']:>3} | "
            f"{s['transport_mean']:>8.4f} {s['transport_std']:>5.4f} | "
            f"{s['baseline_mean']:>8.4f} {s['baseline_std']:>5.4f} | "
            f"{s['improve_mean_pct']:>6.1f}% {s['transport_wins_pct']:>5.1f}%"
        )
    return rows


# --- вызов: подставь pos, neg из torch.load и нужный step ---
# data_pos = torch.load("...", map_location="cpu")
# data_neg = torch.load("...", map_location="cpu")
```

```python
import torch
from src.steering.subspace_act import ContrastivePCATokenWise

# =====================
# CONFIG
# =====================
STEP = 0
BRANCH = "txt"
N_SAMPLES = 100

# "очистка" / concept subspace
SUBSPACE_DIM = 32
MIN_EXPLAINED = 0.90
PCA_DELTA_MODE = "raw_delta"   # centered | raw_delta | append_mean_dir

# сравнение с нейтральным уже в low-d
KMAX_LOW = 64                  # <= d
K_REPORT = [3, 8, 16, 32, 64, 512]


def _to_ntc(x):
    if x.dim() == 4 and x.shape[1] == 1:
        x = x.squeeze(1)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,T,C), got {tuple(x.shape)}")
    return x.float()

def _extract_layer_branch(data, step, layer_key, branch="txt"):
    x = data[step][layer_key]
    if isinstance(x, dict):
        x = x[branch]
    return _to_ntc(x)

def _fit_cpca_like_extraction(neg_ntc, pos_ntc, subspace_dim, min_explained, pca_delta_mode):
    finalize_append = pca_delta_mode == "append_mean_dir"
    cpca = ContrastivePCATokenWise(
        subspace_dim=subspace_dim,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_ntc, pos_ntc)
    evr = cpca.explained_variance_ratio()
    if evr.numel() == 0:
        d_sel = 1
    else:
        d_sel = int((evr.cumsum(0) < min_explained).sum().item()) + 1
        d_sel = max(1, min(d_sel, cpca.U.shape[1]))
    cpca = ContrastivePCATokenWise(
        subspace_dim=d_sel,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_ntc, pos_ntc, finalize_append_mean=finalize_append)
    return cpca

def _fit_neutral_basis_lowd(Z_neutral_ntd, kmax):
    # Z_neutral_ntd: (N,T,d)
    X = Z_neutral_ntd.reshape(-1, Z_neutral_ntd.shape[-1])       # (N*T,d)
    Xc = X - X.mean(dim=0, keepdim=True)
    _, _, Vh = torch.linalg.svd(Xc, full_matrices=False)
    k_eff = min(kmax, Vh.shape[0], Vh.shape[1])
    return Vh[:k_eff].T.contiguous()                              # (d,k_eff)

def _overlap_tokens_vs_k_lowd(delta_tok_d, U_neutral_low, eps=1e-12):
    # delta_tok_d: (T,d), U_neutral_low: (d,K)
    denom = delta_tok_d.norm(dim=1, keepdim=True).clamp(min=eps)
    cols = []
    K = U_neutral_low.shape[1]
    for k in range(1, K + 1):
        Uk = U_neutral_low[:, :k]                                 # (d,k)
        proj = (delta_tok_d @ Uk) @ Uk.T                          # (T,d)
        ov = proj.norm(dim=1, keepdim=True) / denom
        cols.append(ov)
    return torch.cat(cols, dim=1)                                  # (T,K)

layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if isinstance(k, str) and k.startswith("layer_")],
    key=lambda s: int(s.split("_")[1])
)

print(f"=== CLEANED overlap in low-d | step={STEP}, branch={BRANCH}, N={N_SAMPLES} ===\n")

rows = []
for layer_key in layer_keys:
    pos_ntc = _extract_layer_branch(pos_data, STEP, layer_key, BRANCH)
    neg_ntc = _extract_layer_branch(acts_current, STEP, layer_key, BRANCH)

    n = min(N_SAMPLES, pos_ntc.shape[0], neg_ntc.shape[0])
    pos_ntc, neg_ntc = pos_ntc[:n], neg_ntc[:n]

    # 1) concept subspace (очистка) как в extraction
    cpca = _fit_cpca_like_extraction(
        neg_ntc, pos_ntc,
        subspace_dim=SUBSPACE_DIM,
        min_explained=MIN_EXPLAINED,
        pca_delta_mode=PCA_DELTA_MODE,
    )
    Uc = cpca.U.float()                                            # (C,d)
    d = Uc.shape[1]

    # 2) CLEANED steering: delta_tc -> low-d
    delta_tc = (pos_ntc - neg_ntc).mean(dim=0)                     # (T,C)
    #delta_tok_d = delta_tc @ Uc                                    # (T,d)
    delta_clean_tc = (delta_tc @ Uc) @ Uc.T           # (T,C)

    # 3) neutral activations in same low-d
    # Z_neutral_ntd = (neg_ntc - cpca.mu_neg.float()) @ Uc          # (N,T,d)

    # 4) neutral basis in low-d and overlap
    U_neutral_low = _fit_neutral_basis_lowd(neg_ntc.mean(dim=0), KMAX_LOW)  # (d,k_eff)
    ov_map = _overlap_tokens_vs_k_lowd(delta_clean_tc, U_neutral_low)    # (T,k_eff)
    k_eff = ov_map.shape[1]

    auc = float(ov_map.mean())  # mean over tokens and k
    layer_idx = int(layer_key.split("_")[1])

    parts = [f"layer_{layer_idx:02d} | N={n:3d} T={ov_map.shape[0]:3d} d={d:2d} K={k_eff:2d} | AUC={auc:.4f}"]
    k_stats = {}
    for k in K_REPORT:
        if k <= k_eff:
            v = ov_map[:, k-1]
            m = float(v.mean())
            md = float(torch.median(v))
            sd = float(v.std(unbiased=False))
            parts.append(f"k={k:2d}: mean={m:.4f} med={md:.4f} std={sd:.4f}")
            k_stats[k] = m
    print(" || ".join(parts))

    rows.append({"layer": layer_idx, "auc": auc, "k_stats": k_stats})

print("\n=== top layers by AUC (cleaned, low-d) ===")
for r in sorted(rows, key=lambda x: x["auc"], reverse=True)[:5]:
    extras = ", ".join([f"m@{k}={v:.4f}" for k, v in sorted(r["k_stats"].items())])
    print(f"layer_{r['layer']:02d}: AUC={r['auc']:.4f}" + (f" | {extras}" if extras else ""))
```
