## step 0 layers all

```python
import math
# ===== config =====
STEP = 3
BRANCH = "txt"      # "txt" or "img"
N_SAMPLES = 20
D_MAX = 64
CENTER = False      # True => SVD(delta - mu_delta), False => SVD(delta)
NCOLS = 5
pos_data = torch.load(POS_ACT_PATH, map_location="cpu")
neg_data = torch.load(NEG_ACT_PATH, map_location="cpu")
layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if k.startswith("layer_")],
    key=lambda x: int(x.split("_")[1])
)
n_layers = len(layer_keys)
nrows = math.ceil(n_layers / NCOLS)
fig, axes = plt.subplots(nrows, NCOLS, figsize=(NCOLS * 3.4, nrows * 2.8), squeeze=False)
for i, layer_key in enumerate(layer_keys):
    r, c = divmod(i, NCOLS)
    ax = axes[r][c]
    x_pos = pos_data[STEP][layer_key]
    x_neg = neg_data[STEP][layer_key]
    
    if isinstance(x_pos, dict):   # dual branch format
        x_pos = x_pos[BRANCH].mean(dim=0)
        x_neg = x_neg[BRANCH].mean(dim=0)
    
    print(x_pos.shape, x_neg.shape)
    if x_pos.dim() == 4 and x_pos.shape[1] == 1:
        x_pos = x_pos.squeeze(1)
    if x_neg.dim() == 4 and x_neg.shape[1] == 1:
        x_neg = x_neg.squeeze(1)
    n = min(N_SAMPLES, x_pos.shape[0], x_neg.shape[0])
    pos = x_pos[:n].float()
    neg = x_neg[:n].float()
    delta = (pos - neg).reshape(-1, pos.shape[-1])  # (N*T, C)
    mu_delta = delta.mean(dim=0)
    X = delta - mu_delta.unsqueeze(0) if CENTER else delta
    _, S, Vh = torch.linalg.svd(X, full_matrices=False)
    evr = (S.pow(2).cumsum(0) / S.pow(2).sum().clamp(min=1e-12))[:D_MAX]
    d90 = int((evr < 0.90).sum().item()) + 1
    pc1 = Vh[0]
    cos = torch.nn.functional.cosine_similarity(mu_delta.unsqueeze(0), pc1.unsqueeze(0)).item()
    ax.plot(torch.arange(1, len(evr) + 1), evr.numpy(), linewidth=1.5)
    ax.axhline(0.9, linestyle="--", linewidth=1)
    ax.set_ylim(0, 1.02)
    ax.set_title(f"{layer_key}\nd90={d90}, cos={cos:.2f}", fontsize=8)
    ax.grid(alpha=0.25)
# hide empty axes
for j in range(n_layers, nrows * NCOLS):
    r, c = divmod(j, NCOLS)
    axes[r][c].axis("off")
fig.suptitle(f"EVR per layer | step={STEP} | branch={BRANCH} | center={CENTER}", fontsize=12)
plt.tight_layout()
plt.show()
```

## Реконструкционная ошибка vs d

```python
# ==========================================================
# Check 1: Reconstruction error vs d
# ==========================================================
# Expects:
#   pos_acts, neg_acts: torch.Tensor of shape (N, T, C)
# Example:
#   pos_acts = pos_data[step][layer]["txt"].squeeze(1).float()[:N]
#   neg_acts = neg_data[step][layer]["txt"].squeeze(1).float()[:N]

import torch
import matplotlib.pyplot as plt

def reconstruction_error_vs_d(
    pos_acts: torch.Tensor,
    neg_acts: torch.Tensor,
    d_max: int = 50,
    center: bool = False,
):
    """
    Returns:
      ds            : list[int]
      rel_errors    : list[float]
      svals         : torch.Tensor (all singular values)
      delta_ref     : (T, C) mean shift used as target
      delta_centered: matrix actually decomposed by SVD
    """
    assert pos_acts.dim() == 3 and neg_acts.dim() == 3, (pos_acts.shape, neg_acts.shape)
    assert pos_acts.shape == neg_acts.shape, (pos_acts.shape, neg_acts.shape)

    # Mean paired token-wise shift (T, C)
    delta_ref = (pos_acts - neg_acts).mean(dim=0).float()  # (T, C)

    # Matrix used in SVD
    if center:
        row_mean = delta_ref.mean(dim=0, keepdim=True)     # (1, C)
        X = delta_ref - row_mean
    else:
        X = delta_ref

    # SVD: X = U @ diag(S) @ Vh
    U, S, Vh = torch.linalg.svd(X, full_matrices=False)
    r = S.numel()
    d_max_eff = min(int(d_max), int(r))

    denom = delta_ref.norm().clamp(min=1e-12)
    ds, rel_errors = [], []

    for d in range(1, d_max_eff + 1):
        # Rank-d reconstruction of X
        X_approx = (U[:, :d] * S[:d]) @ Vh[:d, :]          # (T, C)

        # Compare against original delta_ref (undo centering if needed)
        if center:
            delta_approx = X_approx + row_mean
        else:
            delta_approx = X_approx

        err = (delta_ref - delta_approx).norm() / denom
        ds.append(d)
        rel_errors.append(float(err))

    return ds, rel_errors, S, delta_ref, X


# ===== Example usage =====
# ds, rel_errors, S, delta_ref, X = reconstruction_error_vs_d(
#     pos_acts, neg_acts, d_max=50, center=False
# )
# plt.figure(figsize=(6,4))
# plt.plot(ds, rel_errors, marker="o", markersize=3)
# plt.xlabel("d")
# plt.ylabel("relative reconstruction error")
# plt.title("Reconstruction error vs d")
# plt.grid(alpha=0.3)
# plt.tight_layout()
# plt.show()
```

```python

def _to_ntc(x):
    if x.dim() == 4 and x.shape[1] == 1:
        x = x.squeeze(1)  # (N, T, C)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,T,C), got {tuple(x.shape)}")
    return x.float()
def _get_branch_tensor(data, step, layer_key, branch="txt"):
    x = data[step][layer_key]
    if isinstance(x, dict):
        x = x[branch]
    return _to_ntc(x)
def layer_metrics(pos_ntc, neg_ntc, d_max=64, center=False):
    # pos_ntc, neg_ntc: (N,T,C)
    n = min(pos_ntc.shape[0], neg_ntc.shape[0], N_SAMPLES)
    pos_ntc = pos_ntc[:n]
    neg_ntc = neg_ntc[:n]
    # Mean paired token-wise shift (T,C)
    delta_ref = (pos_ntc - neg_ntc).mean(dim=0).float()
    if center:
        row_mean = delta_ref.mean(dim=0, keepdim=True)     # (1,C)
        X = delta_ref - row_mean
    else:
        row_mean = None
        X = delta_ref
    # SVD: X = U @ diag(S) @ Vh
    U, S, Vh = torch.linalg.svd(X, full_matrices=False)
    r = int(S.numel())
    d_max_eff = min(int(d_max), r)
    ev = S.pow(2)
    cum_evr = ev.cumsum(0) / ev.sum().clamp(min=1e-12)
    cum_evr = cum_evr[:d_max_eff]
    # cos(mu_delta, PC1)
    mu_delta = delta_ref.mean(dim=0)  # mean over token rows => (C,)
    pc1 = Vh[0]
    cos = torch.nn.functional.cosine_similarity(mu_delta.unsqueeze(0), pc1.unsqueeze(0)).item()
    # reconstruction error vs d (against delta_ref)
    denom = delta_ref.norm().clamp(min=1e-12)
    rel_err = []
    ds = list(range(1, d_max_eff + 1))
    for d in ds:
        X_approx = (U[:, :d] * S[:d]) @ Vh[:d, :]
        if center:
            delta_approx = X_approx + row_mean
        else:
            delta_approx = X_approx
        err = (delta_ref - delta_approx).norm() / denom
        rel_err.append(float(err))
    d90 = int((cum_evr < 0.90).sum().item()) + 1
    return ds, rel_err, cum_evr.cpu(), d90, cos
# ---------- Load ----------
pos_data = torch.load(POS_ACT_PATH, map_location="cpu")
neg_data = torch.load(NEG_ACT_PATH, map_location="cpu")
layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if k.startswith("layer_")],
    key=lambda s: int(s.split("_")[1])
)
# ---------- Plot Grid ----------
n_layers = len(layer_keys)
nrows = math.ceil(n_layers / NCOLS)
fig, axes = plt.subplots(nrows, NCOLS, figsize=(NCOLS * 4.0, nrows * 3.1), squeeze=False)
for idx, layer_key in enumerate(layer_keys):
    r, c = divmod(idx, NCOLS)
    ax = axes[r][c]
    try:
        pos_ntc = _get_branch_tensor(pos_data, STEP, layer_key, branch=BRANCH)
        neg_ntc = _get_branch_tensor(neg_data, STEP, layer_key, branch=BRANCH)
        ds, rel_err, cum_evr, d90, cos = layer_metrics(pos_ntc, neg_ntc, d_max=D_MAX, center=CENTER)
        # left axis: reconstruction error
        ax.plot(ds, rel_err, color="tab:red", linewidth=1.7, label="rel err")
        ax.set_ylim(0.0, 1.05)
        ax.set_xlabel("d", fontsize=8)
        ax.set_ylabel("err", fontsize=8, color="tab:red")
        ax.tick_params(axis="y", labelcolor="tab:red", labelsize=7)
        ax.tick_params(axis="x", labelsize=7)
        ax.grid(alpha=0.25)
        # right axis: EVR
        ax2 = ax.twinx()
        ax2.plot(ds, cum_evr.numpy(), color="tab:blue", linewidth=1.5, label="cum EVR")
        ax2.axhline(0.9, linestyle="--", linewidth=1)
        ax2.set_ylim(0.0, 1.02)
        ax2.set_ylabel("EVR", fontsize=8, color="tab:blue")
        ax2.tick_params(axis="y", labelcolor="tab:blue", labelsize=7)
        ax.set_title(f"{layer_key}\nd90={d90}, cos={cos:.2f}", fontsize=8)
    except Exception as e:
        ax.axis("off")
        ax.set_title(f"{layer_key}\nERROR: {str(e)[:40]}...", fontsize=8)
# hide empty cells
for j in range(n_layers, nrows * NCOLS):
    rr, cc = divmod(j, NCOLS)
    axes[rr][cc].axis("off")
fig.suptitle(
    f"Step={STEP} | branch={BRANCH} | center={CENTER}\n"
    f"red: reconstruction error, blue: cumulative EVR",
    fontsize=12
)
plt.tight_layout()
plt.show()
```

## Устойчивость по промптам

```python

def _to_ntc(x):
    # (N,1,T,C) -> (N,T,C)
    if x.dim() == 4 and x.shape[1] == 1:
        x = x.squeeze(1)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,T,C), got {tuple(x.shape)}")
    return x.float()
def _extract_layer_branch(data, step, layer_key, branch="txt"):
    x = data[step][layer_key]
    if isinstance(x, dict):
        x = x[branch]
    return _to_ntc(x)
def top_dir_cosine_matrix(pos_ntc, neg_ntc, n_samples=20, use_abs=True, center_each_pair=False):
    """
    pos_ntc, neg_ntc: (N,T,C)
    Returns:
      cos_mat: (n,n)
      pair_vals: flattened upper-tri values
    """
    n = min(n_samples, pos_ntc.shape[0], neg_ntc.shape[0])
    pos_ntc = pos_ntc[:n]
    neg_ntc = neg_ntc[:n]
    top_dirs = []
    for k in range(n):
        delta_k = (pos_ntc[k] - neg_ntc[k]).float()   # (T,C)
        if center_each_pair:
            delta_k = delta_k - delta_k.mean(dim=0, keepdim=True)
        # SVD on (T,C), right singular vectors in feature space C
        _, _, Vh_k = torch.linalg.svd(delta_k, full_matrices=False)
        v = Vh_k[0]
        v = v / v.norm().clamp(min=1e-12)
        top_dirs.append(v)
    top_dirs = torch.stack(top_dirs, dim=0)            # (n,C)
    cos_mat = top_dirs @ top_dirs.T                    # normalized => cosine matrix
    if use_abs:
        cos_mat = cos_mat.abs()
    vals = []
    for i in range(n):
        for j in range(i + 1, n):
            vals.append(float(cos_mat[i, j]))
    vals = torch.tensor(vals) if len(vals) else torch.tensor([])
    return cos_mat.cpu(), vals
# ---------- load ----------
USE_ABS_COS = True      # обычно лучше True, т.к. знак сингулярного вектора произволен
CENTER_EACH_PAIR = False

pos_data = torch.load(POS_ACT_PATH, map_location="cpu")
neg_data = torch.load(NEG_ACT_PATH, map_location="cpu")
layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if k.startswith("layer_")],
    key=lambda s: int(s.split("_")[1])
)
# ---------- compute for all layers ----------
cos_by_layer = {}
stats = []  # (layer_idx, median, q25, q75, mean, std)
for layer_key in layer_keys:
    pos_ntc = _extract_layer_branch(pos_data, STEP, layer_key, branch=BRANCH)
    neg_ntc = _extract_layer_branch(neg_data, STEP, layer_key, branch=BRANCH)
    cos_mat, vals = top_dir_cosine_matrix(
        pos_ntc, neg_ntc,
        n_samples=N_SAMPLES,
        use_abs=USE_ABS_COS,
        center_each_pair=CENTER_EACH_PAIR,
    )
    cos_by_layer[layer_key] = cos_mat
    if vals.numel() > 0:
        q25 = float(torch.quantile(vals, 0.25))
        med = float(torch.quantile(vals, 0.50))
        q75 = float(torch.quantile(vals, 0.75))
        mean = float(vals.mean())
        std = float(vals.std(unbiased=False))
    else:
        q25 = med = q75 = mean = std = float("nan")
    layer_idx = int(layer_key.split("_")[1])
    stats.append((layer_idx, med, q25, q75, mean, std))
stats = sorted(stats, key=lambda x: x[0])
# =======================
# 1) Grid heatmap per layer
# =======================
n_layers = len(layer_keys)
nrows = math.ceil(n_layers / NCOLS)
fig, axes = plt.subplots(nrows, NCOLS, figsize=(NCOLS * 3.2, nrows * 3.0), squeeze=False)
for i, layer_key in enumerate(layer_keys):
    r, c = divmod(i, NCOLS)
    ax = axes[r][c]
    cm = cos_by_layer[layer_key]
    im = ax.imshow(cm.numpy(), vmin=0.0 if USE_ABS_COS else -1.0, vmax=1.0, cmap="viridis")
    idx = int(layer_key.split("_")[1])
    # quick stat in title
    vals = cm[torch.triu(torch.ones_like(cm), diagonal=1) > 0]
    med = float(torch.median(vals)) if vals.numel() else float("nan")
    ax.set_title(f"layer_{idx}\nmed={'{:.2f}'.format(med)}", fontsize=8)
    ax.set_xticks([])
    ax.set_yticks([])
for j in range(n_layers, nrows * NCOLS):
    rr, cc = divmod(j, NCOLS)
    axes[rr][cc].axis("off")
cbar = fig.colorbar(im, ax=axes, fraction=0.015, pad=0.02)
cbar.set_label("|cos|" if USE_ABS_COS else "cos", rotation=90)
fig.suptitle(
    f"Top-1 direction stability across prompts | step={STEP}, branch={BRANCH}, abs={USE_ABS_COS}",
    fontsize=12
)
plt.tight_layout()
plt.show()

# =======================
# 2) Summary across layers (median + IQR)
# =======================
layer_ids = [x[0] for x in stats]
medians   = [x[1] for x in stats]
q25s      = [x[2] for x in stats]
q75s      = [x[3] for x in stats]
means     = [x[4] for x in stats]
# asymmetric error bars for IQR
yerr_lower = [m - q for m, q in zip(medians, q25s)]
yerr_upper = [q - m for m, q in zip(medians, q75s)]
plt.figure(figsize=(10, 4))
plt.errorbar(layer_ids, medians, yerr=[yerr_lower, yerr_upper], fmt='o-', capsize=3, label='median ± IQR')
plt.plot(layer_ids, means, '--', alpha=0.7, label='mean')
plt.ylim(0.0 if USE_ABS_COS else -1.0, 1.0)
plt.xlabel("Layer")
plt.ylabel("|cos| pairwise top-dir" if USE_ABS_COS else "cos pairwise top-dir")
plt.title(f"Direction stability by layer | step={STEP}, branch={BRANCH}")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import torch
import matplotlib.pyplot as plt

# neg_acts, pos_acts: (N, T, C) — одна пара архивов, один слой, один step
# N = число промптов (пар)

def pooled_activation_delta(pos_acts, neg_acts):
    """Одна «стрелка» на промпт в полном C: mean_t (a+ - a-)."""
    dlt = (pos_acts.float() - neg_acts.float()).mean(dim=1)  # (N, C)
    return dlt / dlt.norm(dim=1, keepdim=True).clamp(min=1e-8)


def pairwise_cosine_matrix(X):
    """X: (N, D) L2-нормированные строки → косинусы = X @ X.T"""
    return (X @ X.T).cpu()


def plot_cosine_hist(G):
    N = G.shape[0]
    iu = torch.triu_indices(N, N, offset=1)
    vals = G[iu[0], iu[1]].numpy()
    plt.figure(figsize=(6, 3))
    plt.hist(vals, bins=40, color="steelblue", alpha=0.85)
    plt.xlabel("cos(Δ_i, Δ_j)")
    plt.ylabel("count")
    plt.title("Pairwise cosine between per-prompt steering directions")
    plt.tight_layout()
    plt.show()
    print(f"pairwise cos: mean={vals.mean():.3f}, std={vals.std():.3f}, "
          f"min={vals.min():.3f}, max={vals.max():.3f}")


# --- 1) Вектор на промпт ---
Delta = pooled_activation_delta(pos_acts, neg_acts)  # (N, C), нормированы

G = pairwise_cosine_matrix(Delta)
plot_cosine_hist(G)

# --- 2) Кластеризация в R^C (или сначала PCA для скорости) ---
try:
    from sklearn.cluster import KMeans
    from sklearn.decomposition import PCA
    from sklearn.metrics import silhouette_score
except ImportError:
    raise RuntimeError("pip install scikit-learn")

X = Delta.cpu().numpy()
# опционально: сжать перед k-means (быстрее и стабильнее при большом C)
pca_pre = PCA(n_components=min(64, X.shape[1], X.shape[0] - 1), random_state=0)
Xz = pca_pre.fit_transform(X)

for K in [2, 3, 4]:
    if Xz.shape[0] < K + 1:
        continue
    km = KMeans(n_clusters=K, n_init=20, random_state=0)
    lab = km.fit_predict(Xz)
    sil = silhouette_score(Xz, lab)
    print(f"K={K}  silhouette={sil:.3f}  counts={[(lab == j).sum() for j in range(K)]}")

# визуализация кластеров в 2D (PCA от Delta)
pca2 = PCA(n_components=2, random_state=0)
XY = pca2.fit_transform(X)
K_plot = 3
km = KMeans(n_clusters=K_plot, n_init=20, random_state=0)
lab = km.fit_predict(Xz)
plt.figure(figsize=(5, 4))
for j in range(K_plot):
    m = lab == j
    plt.scatter(XY[m, 0], XY[m, 1], s=18, alpha=0.75, label=f"cluster {j}")
plt.legend(markerscale=1.2, fontsize=8)
plt.title("Per-prompt Δ (pooled) — PCA-2D, colors = KMeans")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.tight_layout()
plt.show()
```

```python
import torch

# pos_acts, neg_acts: (N, T, C) для выбранного step/layer/branch
# Пример получения:
# pos_acts = ...
# neg_acts = ...

def fit_U_from_acts(
    pos_acts: torch.Tensor,
    neg_acts: torch.Tensor,
    d: int = 32,
    pca_delta_mode: str = "raw_delta",   # "raw_delta" | "centered" | "append_mean_dir"
):
    """
    Возвращает:
      U      : (C, d_eff)
      mu_neg : (T, C)   (mean over batch, tokenwise)
      S      : сингулярные значения
      Vh     : правые сингулярные векторы (d_full, C)
    """
    assert pos_acts.shape == neg_acts.shape and pos_acts.dim() == 3
    N, T, C = pos_acts.shape

    pos = pos_acts.float()
    neg = neg_acts.float()

    # tokenwise anchor (как ты и хотела: только mean over batch)
    mu_neg = neg.mean(dim=0)                         # (T, C)

    # paired deltas
    delta = pos - neg                                # (N, T, C)
    X = delta.reshape(N * T, C)                      # (N*T, C)

    if pca_delta_mode == "centered":
        X = X - X.mean(dim=0, keepdim=True)
    elif pca_delta_mode == "append_mean_dir":
        m = X.mean(dim=0, keepdim=True)              # (1, C)
        Xc = X - m
        # усиливаем mean direction, чтобы точно попасть в подпространство
        X = torch.cat([Xc, m], dim=0)
    elif pca_delta_mode == "raw_delta":
        pass
    else:
        raise ValueError(f"Unknown pca_delta_mode={pca_delta_mode}")

    # SVD по фичам C
    # X = U_s @ diag(S) @ Vh, где строки Vh — базисные направления в R^C
    _, S, Vh = torch.linalg.svd(X, full_matrices=False)

    d_eff = min(int(d), int(Vh.shape[0]))
    U = Vh[:d_eff].T.contiguous()                    # (C, d_eff)

    return U, mu_neg, S, Vh


# ===== usage =====
U, mu_neg, S, Vh = fit_U_from_acts(
    pos_acts, neg_acts,
    d=32,
    pca_delta_mode="raw_delta",
)
print("U:", U.shape, "mu_neg:", mu_neg.shape)
```

```python
import torch
from src.steering.subspace_act import ContrastivePCATokenWise, AffineTransport


def run_split_eval(Z_neg, Z_pos, train_frac=0.8, seed=42):
    """Z_neg, Z_pos: (N, T, d) — один random split."""
    g = torch.Generator(device=Z_neg.device)
    g.manual_seed(seed)

    N, T, d = Z_neg.shape
    n_train = int(N * train_frac)
    idx = torch.randperm(N, generator=g, device=Z_neg.device)
    train_idx = idx[:n_train]
    val_idx = idx[n_train:]

    Ws_train, bs_train = [], []
    for t in range(T):
        tr = AffineTransport().fit(
            Z_neg[train_idx, t, :],
            Z_pos[train_idx, t, :],
        )
        Ws_train.append(tr.W)
        bs_train.append(tr.b)
    Ws_train = torch.stack(Ws_train)  # (T, d, d)
    bs_train = torch.stack(bs_train)  # (T, d)

    Z_neg_val = Z_neg[val_idx]
    Z_pos_val = Z_pos[val_idx]

    Z_pred = torch.einsum("ntd,tdk->ntk", Z_neg_val, Ws_train) + bs_train.unsqueeze(0)
    mean_diff = (Z_pos[train_idx] - Z_neg[train_idx]).mean(dim=0)
    Z_pred_baseline = Z_neg_val + mean_diff.unsqueeze(0)

    err_transport_per_sample = (Z_pred - Z_pos_val).norm(dim=-1).mean(dim=-1)
    err_baseline_per_sample = (Z_pred_baseline - Z_pos_val).norm(dim=-1).mean(dim=-1)

    return {
        "err_transport": err_transport_per_sample.mean().item(),
        "err_baseline": err_baseline_per_sample.mean().item(),
        "improve_pct": (
            (err_baseline_per_sample.mean() - err_transport_per_sample.mean())
            / (err_baseline_per_sample.mean() + 1e-12)
            * 100
        ).item(),
    }


def eval_layer_repeated(
    neg_acts,
    pos_acts,
    *,
    subspace_dim=5,
    pca_delta_mode="centered",
    train_frac=0.8,
    num_splits=20,
    seed0=1000,
    min_n=4,
):
    """
    neg_acts, pos_acts: (N, T, C) на одном слое.
    CPCA fit на всех N (как при сохранении вектора); split только для affine / baseline.
    """
    n = min(neg_acts.shape[0], pos_acts.shape[0])
    if n < min_n:
        return None
    neg_acts = neg_acts[:n].float()
    pos_acts = pos_acts[:n].float()

    cpca = ContrastivePCATokenWise(
        subspace_dim=subspace_dim,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_acts, pos_acts)

    Z_neg = cpca.project(neg_acts)
    Z_pos = cpca.project(pos_acts)

    outs = [
        run_split_eval(Z_neg, Z_pos, train_frac=train_frac, seed=seed0 + s)
        for s in range(num_splits)
    ]
    te = torch.tensor([o["err_transport"] for o in outs])
    be = torch.tensor([o["err_baseline"] for o in outs])
    im = torch.tensor([o["improve_pct"] for o in outs])
    return {
        "d": int(cpca.U.shape[1]),
        "T": int(Z_neg.shape[1]),
        "N": int(Z_neg.shape[0]),
        "transport_mean": te.mean().item(),
        "transport_std": te.std(unbiased=True).item(),
        "baseline_mean": be.mean().item(),
        "baseline_std": be.std(unbiased=True).item(),
        "improve_mean_pct": im.mean().item(),
        "improve_std_pct": im.std(unbiased=True).item(),
        "transport_wins_pct": (te < be).float().mean().item() * 100.0,
    }


def eval_all_layers_one_step(
    pos,
    neg,
    step,
    *,
    blocks=19,
    subspace_dim=32,
    pca_delta_mode="centered",
    num_splits=20,
    train_frac=0.8,
):
    rows = []
    for block in range(blocks):
        layer = f"layer_{block}"
        if step not in pos or step not in neg:
            continue
        if layer not in pos[step] or layer not in neg[step]:
            continue
        pn = pos[step][layer]
        nn = neg[step][layer]
        # при dual: pos[step][layer]['txt'] и т.д.
        if isinstance(pn, dict):
            if "txt" in pn:
                pn, nn = pn["txt"], nn["txt"]
            elif "img" in pn:
                pn, nn = pn["img"], nn["img"]
            else:
                continue
        if not (torch.is_tensor(pn) and torch.is_tensor(nn)):
            continue
        while pn.dim() > 3:
            pn = pn.squeeze(1)
            nn = nn.squeeze(1)
        if pn.dim() != 3 or nn.dim() != 3:
            continue

        stats = eval_layer_repeated(
            nn,
            pn,
            subspace_dim=subspace_dim,
            pca_delta_mode=pca_delta_mode,
            train_frac=train_frac,
            num_splits=num_splits,
        )
        if stats is None:
            continue
        rows.append((layer, stats))

    # печать таблицей
    print(f"step={step}  subspace_dim={subspace_dim}  pca_delta_mode={pca_delta_mode}  num_splits={num_splits}\n")
    hdr = f"{'layer':<12} {'N':>4} {'T':>4} {'d':>3} | {'tr_err':>8} {'±':>5} | {'base':>8} {'±':>5} | {'imp%':>7} {'win%':>6}"
    print(hdr)
    print("-" * len(hdr))
    for layer, s in rows:
        print(
            f"{layer:<12} {s['N']:>4} {s['T']:>4} {s['d']:>3} | "
            f"{s['transport_mean']:>8.4f} {s['transport_std']:>5.4f} | "
            f"{s['baseline_mean']:>8.4f} {s['baseline_std']:>5.4f} | "
            f"{s['improve_mean_pct']:>6.1f}% {s['transport_wins_pct']:>5.1f}%"
        )
    return rows


# --- вызов: подставь pos, neg из torch.load и нужный step ---
# data_pos = torch.load("...", map_location="cpu")
# data_neg = torch.load("...", map_location="cpu")
```

```python
import torch
import matplotlib.pyplot as plt
from src.steering.subspace_act import ContrastivePCA, ContrastivePCATokenWise

# =====================
# CONFIG
# =====================
STEP = 0
BRANCH = "txt"
N_SAMPLES = 100
TRAIN_FRAC = 0.8

# Откуда брать текстовые активации:
#   "sequence" — только тензор (N,T,C) или txt["sequence"]
#   "pooled"   — только txt["pooled"] (N,C) → внутри как (N,1,C)
#   "auto"     — если в txt есть "pooled", берём его, иначе sequence / тензор
TXT_SOURCE = "auto"

# Как фитить subspace (как в extraction):
#   "auto"          — если данные из pooled, ContrastivePCA; иначе ContrastivePCATokenWise
#   "tokenwise"   — всегда ContrastivePCATokenWise (нужен (N,T,C), T>=1)
#   "sample_pooled" — всегда ContrastivePCA (mean по токенам на образец)
SUBSPACE_FIT = "auto"

SUBSPACE_DIM = 32
MIN_EXPLAINED = 0.90
PCA_DELTA_MODE = "raw_delta"

EPS = 1e-6
MONGE_DIM_CAP = 8  # можно поставить 2-3 для быстрой визуальной проверки гауссовости
MAKE_SCATTER_FOR_TOP = True


def _to_ntc(x):
    if x.dim() == 4 and x.shape[1] == 1:
        x = x.squeeze(1)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,T,C), got {tuple(x.shape)}")
    return x.float()


def _as_ntc_from_pooled(pooled_nc):
    """(N,C) или (N,1,C) → (N,1,C) float."""
    x = pooled_nc.float()
    if x.dim() == 2:
        x = x.unsqueeze(1)
    return _to_ntc(x)


def _extract_layer_branch(data, step, layer_key, branch="txt", txt_source="auto"):
    """
    Возвращает (neg_or_pos_ntc, meta) где meta["used_pooled"]=True если взяли txt['pooled'].
    """
    layer = data[step][layer_key]
    if not isinstance(layer, dict):
        return _to_ntc(layer), {"used_pooled": False}

    if branch not in layer:
        raise KeyError(f"{layer_key}: нет ключа branch={branch!r}, есть {list(layer.keys())}")

    b = layer[branch]
    used_pooled = False

    if torch.is_tensor(b):
        return _to_ntc(b), {"used_pooled": False}

    if not isinstance(b, dict):
        raise TypeError(f"{layer_key}[{branch}]: ожидали tensor или dict, got {type(b)}")

    if txt_source == "pooled":
        if "pooled" not in b:
            raise KeyError(f"{layer_key}[{branch}]: TXT_SOURCE=pooled, но нет ключа 'pooled'")
        return _as_ntc_from_pooled(b["pooled"]), {"used_pooled": True}

    if txt_source == "sequence":
        if "sequence" in b:
            return _to_ntc(b["sequence"]), {"used_pooled": False}
        # один тензор под другим ключом — не поддерживаем здесь
        raise KeyError(f"{layer_key}[{branch}]: TXT_SOURCE=sequence, нужен ключ 'sequence'")

    # auto
    if "pooled" in b and torch.is_tensor(b["pooled"]):
        return _as_ntc_from_pooled(b["pooled"]), {"used_pooled": True}
    if "sequence" in b and torch.is_tensor(b["sequence"]):
        return _to_ntc(b["sequence"]), {"used_pooled": False}

    raise KeyError(
        f"{layer_key}[{branch}]: auto — ожидал 'pooled' или 'sequence', есть {list(b.keys())}"
    )


def _fit_cpca_like_extraction(neg_ntc, pos_ntc, subspace_dim, min_explained, pca_delta_mode, tokenwise: bool):
    finalize_append = pca_delta_mode == "append_mean_dir"
    CpcaCls = ContrastivePCATokenWise if tokenwise else ContrastivePCA
    cpca = CpcaCls(
        subspace_dim=subspace_dim,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_ntc, pos_ntc)

    evr = cpca.explained_variance_ratio()
    if evr.numel() == 0:
        d_sel = 1
    else:
        d_sel = int((evr.cumsum(0) < min_explained).sum().item()) + 1
        d_sel = max(1, min(d_sel, cpca.U.shape[1]))

    cpca = CpcaCls(
        subspace_dim=d_sel,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_ntc, pos_ntc, finalize_append_mean=finalize_append)
    return cpca


def _mu_anchor_for_pooled_z(neg_ntc, cpca, tokenwise: bool):
    """
    Для Monge мы строим z_k из mean по токенам: ā_k ∈ R^C.
    - ContrastivePCA: cpca.mu_neg уже mean_{n,t}(neg) — совпадает с глобальным якорем.
    - TokenWise: cpca.mu_neg — (T,C); для ā_k используем глобальный якорь mean_{n,t}(neg).
    """
    if not tokenwise:
        return cpca.mu_neg.float()
    return neg_ntc.mean(dim=(0, 1)).float()


def _covariance(z_nd, eps=EPS):
    zc = z_nd - z_nd.mean(dim=0, keepdim=True)
    cov = (zc.T @ zc) / max(1, z_nd.shape[0])
    cov = cov + eps * torch.eye(cov.shape[0], device=cov.device, dtype=cov.dtype)
    return cov


def _sqrt_and_invsqrt_spd(S, eps=EPS):
    vals, vecs = torch.linalg.eigh(S)
    vals = vals.clamp(min=eps)
    S_sqrt = vecs @ torch.diag(vals.sqrt()) @ vecs.T
    S_invsqrt = vecs @ torch.diag(vals.rsqrt()) @ vecs.T
    return S_sqrt, S_invsqrt


def _monge_matrix(Sigma_neg, Sigma_pos, eps=EPS):
    S_neg_sqrt, S_neg_invsqrt = _sqrt_and_invsqrt_spd(Sigma_neg, eps=eps)
    M_mid = S_neg_sqrt @ Sigma_pos @ S_neg_sqrt
    vals_mid, vecs_mid = torch.linalg.eigh(M_mid)
    vals_mid = vals_mid.clamp(min=eps)
    M_mid_sqrt = vecs_mid @ torch.diag(vals_mid.sqrt()) @ vecs_mid.T
    M = S_neg_invsqrt @ M_mid_sqrt @ S_neg_invsqrt
    return M


def _skew_kurtosis_proxy(z_nd, eps=1e-8):
    zc = z_nd - z_nd.mean(dim=0, keepdim=True)
    std = zc.std(dim=0, unbiased=False).clamp(min=eps)
    zn = zc / std
    skew = (zn ** 3).mean(dim=0).abs().mean().item()
    kurt = ((zn ** 4).mean(dim=0) - 3.0).abs().mean().item()
    return skew, kurt


layer_keys = sorted(
    [k for k in pos_data[STEP].keys() if isinstance(k, str) and k.startswith("layer_")],
    key=lambda s: int(s.split("_")[1]),
)

results = []
payload_by_layer = {}

print(
    f"=== Monge diagnostics | step={STEP}, branch={BRANCH}, N={N_SAMPLES} | "
    f"TXT_SOURCE={TXT_SOURCE} SUBSPACE_FIT={SUBSPACE_FIT} ===\n"
)

for layer_key in layer_keys:
    pos_ntc, pos_meta = _extract_layer_branch(pos_data, STEP, layer_key, BRANCH, TXT_SOURCE)
    neg_ntc, neg_meta = _extract_layer_branch(acts_current, STEP, layer_key, BRANCH, TXT_SOURCE)
    if pos_meta != neg_meta:
        print(f"[skip {layer_key}] pos/neg txt layout mismatch: {pos_meta} vs {neg_meta}")
        continue

    used_pooled = pos_meta["used_pooled"]
    if SUBSPACE_FIT == "auto":
        tokenwise = not used_pooled
    elif SUBSPACE_FIT == "tokenwise":
        tokenwise = True
    else:
        tokenwise = False

    n = min(N_SAMPLES, pos_ntc.shape[0], neg_ntc.shape[0])
    if n < 6:
        continue

    pos_ntc = pos_ntc[:n]
    neg_ntc = neg_ntc[:n]

    n_train = max(4, int(n * TRAIN_FRAC))
    n_train = min(n_train, n - 2)

    pos_tr, pos_va = pos_ntc[:n_train], pos_ntc[n_train:n]
    neg_tr, neg_va = neg_ntc[:n_train], neg_ntc[n_train:n]

    cpca = _fit_cpca_like_extraction(
        neg_tr,
        pos_tr,
        subspace_dim=SUBSPACE_DIM,
        min_explained=MIN_EXPLAINED,
        pca_delta_mode=PCA_DELTA_MODE,
        tokenwise=tokenwise,
    )

    V_d = cpca.U.float()  # (C,d)
    d_eff = min(V_d.shape[1], MONGE_DIM_CAP)
    V_d = V_d[:, :d_eff].contiguous()

    mu_neg_c = _mu_anchor_for_pooled_z(neg_tr, cpca, tokenwise)

    # Пулинг по токенам -> N точек в R^d (для pooled-тензора T=1)
    z_neg_tr = (neg_tr.mean(dim=1) - mu_neg_c) @ V_d
    z_pos_tr = (pos_tr.mean(dim=1) - mu_neg_c) @ V_d
    z_neg_va = (neg_va.mean(dim=1) - mu_neg_c) @ V_d
    z_pos_va = (pos_va.mean(dim=1) - mu_neg_c) @ V_d

    mu_neg_z = z_neg_tr.mean(dim=0)
    mu_pos_z = z_pos_tr.mean(dim=0)

    Sigma_neg = _covariance(z_neg_tr, eps=EPS)
    Sigma_pos = _covariance(z_pos_tr, eps=EPS)
    M = _monge_matrix(Sigma_neg, Sigma_pos, eps=EPS)

    # Метрика различия форм облаков
    rel_shape_diff = ((Sigma_pos - Sigma_neg).norm() / Sigma_neg.norm().clamp(min=EPS)).item()

    # A: mean-diff baseline
    delta_mu = mu_pos_z - mu_neg_z
    z_pred_mean = z_neg_va + delta_mu

    # C: Monge map
    z_pred_monge = mu_pos_z + (z_neg_va - mu_neg_z) @ M.T

    # Парная метрика, как в задаче: ||T(z_neg_val) - z_pos_val||
    err_mean = (z_pred_mean - z_pos_va).norm(dim=1).mean().item()
    err_monge = (z_pred_monge - z_pos_va).norm(dim=1).mean().item()

    # Диагностика "насколько гауссово" (близость skew/kurt к 0)
    skew_neg, kurt_neg = _skew_kurtosis_proxy(z_neg_tr)
    skew_pos, kurt_pos = _skew_kurtosis_proxy(z_pos_tr)

    layer_idx = int(layer_key.split("_")[1])
    print(
        f"layer_{layer_idx:02d} | d={d_eff:2d} Ntr={n_train:3d} Nva={n-n_train:3d} | "
        f"shape_diff={rel_shape_diff:.3f} | "
        f"CV mean-diff={err_mean:.4f} vs monge={err_monge:.4f} "
        f"(gain={err_mean - err_monge:+.4f}) | "
        f"gauss_proxy neg(s={skew_neg:.3f},k={kurt_neg:.3f}) pos(s={skew_pos:.3f},k={kurt_pos:.3f})"
    )

    results.append({
        "layer": layer_idx,
        "d": d_eff,
        "n_train": n_train,
        "n_val": n - n_train,
        "shape_diff": rel_shape_diff,
        "cv_mean_diff": err_mean,
        "cv_monge": err_monge,
        "cv_gain": err_mean - err_monge,
        "gauss_neg_skew": skew_neg,
        "gauss_neg_kurt": kurt_neg,
        "gauss_pos_skew": skew_pos,
        "gauss_pos_kurt": kurt_pos,
    })

    payload_by_layer[layer_key] = {
        "method": "monge",
        "M": M.detach().cpu(),
        "mu_neg_z": mu_neg_z.detach().cpu(),
        "mu_pos_z": mu_pos_z.detach().cpu(),
        "V_d": V_d.detach().cpu(),
        "mu_neg": mu_neg_c.detach().cpu(),
    }

print("\n=== Top layers by Monge gain (higher is better) ===")
for r in sorted(results, key=lambda x: x["cv_gain"], reverse=True)[:8]:
    print(
        f"layer_{r['layer']:02d}: gain={r['cv_gain']:+.4f} | "
        f"mean-diff={r['cv_mean_diff']:.4f} -> monge={r['cv_monge']:.4f} | "
        f"shape_diff={r['shape_diff']:.3f}"
    )

# Быстрая визуализация (если d>=2): слой с лучшим gain
if MAKE_SCATTER_FOR_TOP and len(results) > 0:
    best = sorted(results, key=lambda x: x["cv_gain"], reverse=True)[0]
    k = f"layer_{best['layer']}"

    pos_ntc = _extract_layer_branch(pos_data, STEP, k, BRANCH)[:N_SAMPLES]
    neg_ntc = _extract_layer_branch(acts_current, STEP, k, BRANCH)[:N_SAMPLES]

    cpca = _fit_cpca_like_extraction(
        neg_ntc,
        pos_ntc,
        subspace_dim=SUBSPACE_DIM,
        min_explained=MIN_EXPLAINED,
        pca_delta_mode=PCA_DELTA_MODE,
    )
    V_d = cpca.U.float()[:, : max(2, min(cpca.U.shape[1], MONGE_DIM_CAP))]
    mu_neg_c = cpca.mu_neg.float()

    z_neg = (neg_ntc.mean(dim=1) - mu_neg_c) @ V_d
    z_pos = (pos_ntc.mean(dim=1) - mu_neg_c) @ V_d

    if z_neg.shape[1] >= 2:
        plt.figure(figsize=(6, 6))
        plt.scatter(z_neg[:, 0].cpu(), z_neg[:, 1].cpu(), alpha=0.65, label="neg")
        plt.scatter(z_pos[:, 0].cpu(), z_pos[:, 1].cpu(), alpha=0.65, label="pos")
        plt.title(f"Layer {best['layer']:02d}: neg vs pos in low-d")
        plt.xlabel("z1")
        plt.ylabel("z2")
        plt.legend()
        plt.grid(True, alpha=0.2)
        plt.show()

print("\nMonge payload dictionary ready: payload_by_layer['layer_i']")
```

```python
def _to_ntc(x):
    if x.dim() == 4 and x.shape[1] == 1:
        x = x.squeeze(1)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,T,C), got {tuple(x.shape)}")
    return x.float()
def _extract_layer_branch(data, step, layer_key, branch="txt"):
    x = data[step][layer_key]
    if isinstance(x, dict):
        x = x[branch]
    return _to_ntc(x)
```

```python
def _calculate_concept_subspace_ratio(to_modify, payload):
    U = payload["U"].to(to_modify.device).float()       # (C, d)
    mu_neg = payload["mu_neg"].to(to_modify.device).float()  # (C,)
    
    flat = to_modify.reshape(-1, to_modify.shape[-1]).float()  # (T, C)
    centered = flat - mu_neg                                    # (T, C)
    
    # Проекция на концептное подпространство
    z = centered @ U                                    # (T, d)
    proj = z @ U.T                                      # (T, C) — компонента в span(U)
    
    # Норма проекции и норма исходного вектора
    proj_norm = proj.norm(dim=-1)                       # (T,)
    total_norm = centered.norm(dim=-1).clamp(min=1e-8)  # (T,)
    
    # Score для каждого токена
    token_scores = proj_norm / total_norm               # (T,) — от 0 до 1
    
    # Агрегируем по токенам — среднее
    score = token_scores.mean()                         # скаляр
    
    return score, token_scores 
```

```python
def _calculate_concept_subspace_ratio(to_modify, payload):
    """
    to_modify: (N, T, C) or (T, C)
    returns:
      score: scalar (mean over batch and tokens)
      token_scores: (N, T) if input was batched, else (T,)
    """
    U = payload["U"].to(to_modify.device).float()            # (C, d)
    mu_neg = payload["mu_neg"].to(to_modify.device).float()  # (C,)

    # accept both unbatched and batched
    was_unbatched = (to_modify.dim() == 2)
    if was_unbatched:
        to_modify = to_modify.unsqueeze(0)  # -> (1, T, C)

    if to_modify.dim() != 3:
        raise ValueError(f"Expected (N,T,C) or (T,C), got {tuple(to_modify.shape)}")

    x = to_modify.float()  
    #print(x.shape, mu_neg.shape)                                   # (N, T, C)
    centered = x #- mu_neg.unsqueeze(0)             # (N, T, C)

    # projection to concept subspace
    z = centered @ U                                          # (N, T, d)
    proj = z @ U.T                                            # (N, T, C)

    proj_norm = proj.norm(dim=-1)                             # (N, T)
    total_norm = centered.norm(dim=-1).clamp(min=1e-8)        # (N, T)
    token_scores = proj_norm / total_norm                     # (N, T), in [0,1]

    # global mean across batch and tokens
    #print(token_scores.shape)
    score = token_scores.mean()                               # scalar

    if was_unbatched:
        token_scores = token_scores.squeeze(0)                # (T,)

    return score, token_scores
```

```python
def _concept_presence_score(to_modify, payload):
    U        = payload["U"].to(to_modify.device).float()          # (C, d)
    mu_neg   = payload["mu_neg"].to(to_modify.device).float()     # (C,)
    mu_neg_z = payload["z_neg_mean"].to(to_modify.device).float() # (d,)
    mu_pos_z = payload["z_pos_mean"].to(to_modify.device).float() # (d,)

    # направление концептного сдвига в subspace
    delta_z      = mu_pos_z - mu_neg_z                            # (d,)
    delta_z_unit = delta_z / delta_z.norm().clamp(min=1e-8)       # (d,)

    was_unbatched = to_modify.dim() == 2
    if was_unbatched:
        to_modify = to_modify.unsqueeze(0)

    x        = to_modify.float()                                  # (N, T, C)
    centered = x - mu_neg                                         # (N, T, C)
    z        = centered @ U                                       # (N, T, d)

    # проекция на направление концептного сдвига — со знаком
    score_tok = z @ delta_z_unit                                  # (N, T) со знаком
    score     = score_tok.mean()                                  # скаляр

    if was_unbatched:
        score_tok = score_tok.squeeze(0)

    return score, score_tok
```

```python
def _concept_presence_score(to_modify, payload):
    """
    Signed concept-presence score along the concept shift direction in z-space.

    to_modify: (N, T, C) or (T, C)
    returns:
      score: scalar (mean over batch and tokens)
      score_tok: (N, T) if batched else (T,)
    """
    U        = payload["U"].to(to_modify.device).float()          # (C, d)
    mu_neg   = payload["mu_neg"].to(to_modify.device).float()     # (C,)
    mu_neg_z = payload["z_neg_mean"].to(to_modify.device).float() # (d,)
    mu_pos_z = payload["z_pos_mean"].to(to_modify.device).float() # (d,)

    # concept shift direction in subspace
    delta_z = mu_pos_z - mu_neg_z                                 # (d,)
    delta_z_unit = delta_z / delta_z.norm().clamp(min=1e-8)        # (d,)
    #print(delta_z_unit.shape)
    was_unbatched = (to_modify.dim() == 2)
    if was_unbatched:
        to_modify = to_modify.unsqueeze(0)                         # (1, T, C)

    if to_modify.dim() != 3:
        raise ValueError(f"Expected (N,T,C) or (T,C), got {tuple(to_modify.shape)}")

    x = to_modify.float()                                          # (N, T, C)
    centered = x - mu_neg.unsqueeze(0)                      # (N, T, C)
    z = centered @ U
    #print(x.shape)                                               # (N, T, d)

    # signed projection onto concept-shift direction
    # option A (simple)
    score_tok = (z * delta_z_unit.unsqueeze(0)).sum(dim=-1)   # (N, T)                                  # (N, T)
    #print(score_tok.shape)
    #assert False
    score = score_tok.mean()                                       # scalar

    if was_unbatched:
        score_tok = score_tok.squeeze(0)                           # (T,)

    return score, score_tok
```

```python
def _concept_distance_score(to_modify, payload):
    U        = payload["U"].to(to_modify.device).float()          # (C, d)
    mu_neg   = payload["mu_neg"].to(to_modify.device).float()     # (C,)
    mu_neg_z = payload["z_neg_mean"].to(to_modify.device).float() # (d,)
    mu_pos_z = payload["z_pos_mean"].to(to_modify.device).float() # (d,)

    was_unbatched = to_modify.dim() == 2
    if was_unbatched:
        to_modify = to_modify.unsqueeze(0)

    x        = to_modify.float()              # (N, T, C)
    centered = x - mu_neg                     # (N, T, C)
    z        = centered @ U                   # (N, T, d)

    # расстояние до центра нейтрального облака
    d_to_neg = (z - mu_neg_z).norm(dim=-1)   # (N, T)
    # расстояние до центра концептного облака
    d_to_pos = (z - mu_pos_z).norm(dim=-1)   # (N, T)

    # score ∈ [0, 1]
    # score > 0.5 → ближе к нейтральному
    # score < 0.5 → ближе к концептному
    score = d_to_neg / (d_to_neg + d_to_pos + 1e-8)  # (N, T)
    score_mean = score.mean()

    if was_unbatched:
        score = score.squeeze(0)

    return score_mean, score
```

```python
def _concept_distance_score(to_modify, payload):
    """
    Distance-based score between neg/pos centers in z-space.

    to_modify: (N, T, C) or (T, C)
    returns:
      score_mean: scalar (mean over batch and tokens)
      score_tok: (N, T) if batched else (T,)
    """
    U        = payload["U"].to(to_modify.device).float()          # (C, d)
    mu_neg   = payload["mu_neg"].to(to_modify.device).float()     # (C,)
    mu_neg_z = payload["z_neg_mean"].to(to_modify.device).float() # (d,) or (T,d)
    mu_pos_z = payload["z_pos_mean"].to(to_modify.device).float() # (d,) or (T,d)

    was_unbatched = (to_modify.dim() == 2)
    if was_unbatched:
        to_modify = to_modify.unsqueeze(0)                        # (1, T, C)

    if to_modify.dim() != 3:
        raise ValueError(f"Expected (N,T,C) or (T,C), got {tuple(to_modify.shape)}")

    x = to_modify.float()                                         # (N, T, C)
    centered = x - mu_neg.unsqueeze(0)                         # (N, T, C)
    z = centered @ U                                              # (N, T, d)

    # broadcast centers to (1,T,d) or (1,1,d)
    if mu_neg_z.dim() == 1:
        mu_neg_z_b = mu_neg_z.view(1, 1, -1)                      # (1,1,d)
    elif mu_neg_z.dim() == 2:
        mu_neg_z_b = mu_neg_z.view(1, mu_neg_z.shape[0], mu_neg_z.shape[1])  # (1,T,d)
    else:
        raise ValueError(f"mu_neg_z must be (d,) or (T,d), got {tuple(mu_neg_z.shape)}")

    if mu_pos_z.dim() == 1:
        mu_pos_z_b = mu_pos_z.view(1, 1, -1)                      # (1,1,d)
    elif mu_pos_z.dim() == 2:
        mu_pos_z_b = mu_pos_z.view(1, mu_pos_z.shape[0], mu_pos_z.shape[1])  # (1,T,d)
    else:
        raise ValueError(f"mu_pos_z must be (d,) or (T,d), got {tuple(mu_pos_z.shape)}")

    
    d_to_neg = (z - mu_neg_z_b).norm(dim=-1)                      # (N, T)
    d_to_pos = (z - mu_pos_z_b).norm(dim=-1)                      # (N, T)

    score_tok = d_to_neg / (d_to_neg + d_to_pos + 1e-8)           # (N, T)
    score_mean = score_tok.mean()                                 # scalar

    if was_unbatched:
        score_tok = score_tok.squeeze(0)                          # (T,)

    return score_mean, score_tok
```

```python
import torch
import pandas as pd



def compute_scores_per_step_layer(pos_data, neg_data, steering_final, branch="txt"):
    rows = []
    neg_scores = {}
    pos_scores = {}
    delta_scores = {}

    common_steps = [0,1,2,3] #sorted(set(pos_data.keys()) & set(neg_data.keys()) & set(steering_final.keys()))

    for step in common_steps:
        # common layer keys for this step
        pos_layers = {k for k in pos_data[step].keys() if isinstance(k, str) and k.startswith("layer_")}
        neg_layers = {k for k in neg_data[step].keys() if isinstance(k, str) and k.startswith("layer_")}
        st_layers  = {k for k in steering_final[step].keys() if isinstance(k, str) and k.startswith("layer_")}
        common_layers = sorted(pos_layers & neg_layers & st_layers, key=lambda s: int(s.split("_")[1]))

        neg_scores[step] = {}
        pos_scores[step] = {}
        delta_scores[step] = {}

        for layer_key in common_layers:
            # activations
            neg_ntc = _extract_layer_branch(neg_data, step, layer_key, branch=branch)  # (N,T,C)
            pos_ntc = _extract_layer_branch(pos_data, step, layer_key, branch=branch)  # (N,T,C)

            # steering payload (supports both [layer] and [layer][branch] shapes)
            payload_obj = steering_final[step][layer_key]
            if isinstance(payload_obj, dict) and "U" in payload_obj:
                payload = payload_obj
            else:
                payload = payload_obj[branch]

            neg_score, _ = _concept_distance_score(neg_ntc, payload)
            pos_score, _ = _concept_distance_score(pos_ntc, payload)

            neg_v = float(neg_score.item())
            pos_v = float(pos_score.item())
            delta_v = pos_v - neg_v

            neg_scores[step][layer_key] = neg_v
            pos_scores[step][layer_key] = pos_v
            delta_scores[step][layer_key] = delta_v

            rows.append({
                "step": step,
                "layer": int(layer_key.split("_")[1]),
                "layer_key": layer_key,
                "neg_score": neg_v,
                "pos_score": pos_v,
                "delta": delta_v,
            })

    df = pd.DataFrame(rows).sort_values(["step", "layer"]).reset_index(drop=True)
    return df, neg_scores, pos_scores, delta_scores
```

```python
import torch
from pathlib import Path
from src.steering.subspace_act import ContrastivePCA

# --- ваши словари только с pooled (N,C) или (N,1,C) ---
# pos_dict, neg_dict = ...

N_SAMPLES = None
TRAIN_FRAC = 0.8
SUBSPACE_DIM = 32
MIN_EXPLAINED = 0.90
PCA_DELTA_MODE = "raw_delta"
EPS = 1e-6
MONGE_DIM_CAP = 8

SAVE_PATH = None  # например Path("monge_pooled_vectors.pt") или None — не сохранять


def _pooled_to_ntc(pooled):
    x = pooled.float()
    if x.dim() == 2:
        x = x.unsqueeze(1)
    if x.dim() != 3:
        raise ValueError(f"pooled: жду (N,C) или (N,1,C), got {tuple(x.shape)}")
    return x


def _fit_cpca_pooled(neg_ntc, pos_ntc, subspace_dim, min_explained, pca_delta_mode):
    finalize_append = pca_delta_mode == "append_mean_dir"
    cpca = ContrastivePCA(subspace_dim=subspace_dim, pca_delta_mode=pca_delta_mode).fit(
        neg_ntc, pos_ntc
    )
    evr = cpca.explained_variance_ratio()
    d_sel = 1 if evr.numel() == 0 else int((evr.cumsum(0) < min_explained).sum().item()) + 1
    d_sel = max(1, min(d_sel, cpca.U.shape[1]))
    return ContrastivePCA(subspace_dim=d_sel, pca_delta_mode=pca_delta_mode).fit(
        neg_ntc, pos_ntc, finalize_append_mean=finalize_append
    )


def _covariance(z_nd, eps=EPS):
    zc = z_nd - z_nd.mean(dim=0, keepdim=True)
    cov = (zc.T @ zc) / max(1, z_nd.shape[0])
    return cov + eps * torch.eye(cov.shape[0], device=cov.device, dtype=cov.dtype)


def _monge_matrix(Sigma_neg, Sigma_pos, eps=EPS):
    vals, vecs = torch.linalg.eigh(Sigma_neg)
    vals = vals.clamp(min=eps)
    S_sqrt = vecs @ torch.diag(vals.sqrt()) @ vecs.T
    S_invsqrt = vecs @ torch.diag(vals.rsqrt()) @ vecs.T
    M_mid = S_sqrt @ Sigma_pos @ S_sqrt
    vm, em = torch.linalg.eigh(M_mid)
    vm = vm.clamp(min=eps)
    M_mid_sqrt = em @ torch.diag(vm.sqrt()) @ em.T
    return S_invsqrt @ M_mid_sqrt @ S_invsqrt


def _z_coords(h_nc, mu_c, V_cd):
    """h: (N,C), mu: (C,), V: (C,d) -> z: (N,d)"""
    return (h_nc - mu_c) @ V_cd


def _lift_z_to_c(delta_z_nd, V_cd):
    """Столбцы U ортонормированы (SVD) → приближённый подъём в C: Δh ≈ Δz @ V.T"""
    return delta_z_nd @ V_cd.T


# -------- 1) данные pooled -> (N,1,C) --------
pos_ntc = _pooled_to_ntc(pos_dict["pooled"])
neg_ntc = _pooled_to_ntc(neg_dict["pooled"])
if pos_ntc.shape != neg_ntc.shape:
    raise ValueError(f"shape pos {tuple(pos_ntc.shape)} vs neg {tuple(neg_ntc.shape)}")

n = pos_ntc.shape[0]
if N_SAMPLES is not None:
    n = min(n, int(N_SAMPLES))
if n < 6:
    raise ValueError(f"Need at least 6 paired samples, got {n}")
pos_ntc, neg_ntc = pos_ntc[:n], neg_ntc[:n]

h_pos = pos_ntc.mean(dim=1)   # (N, C)
h_neg = neg_ntc.mean(dim=1)   # (N, C)
C = h_pos.shape[1]

n_train = max(4, int(n * TRAIN_FRAC))
n_train = min(n_train, n - 2)
pos_tr, pos_va = pos_ntc[:n_train], pos_ntc[n_train:n]
neg_tr, neg_va = neg_ntc[:n_train], neg_ntc[n_train:n]
h_pos_tr, h_neg_tr = pos_tr.mean(dim=1), neg_tr.mean(dim=1)
h_pos_va, h_neg_va = pos_va.mean(dim=1), neg_va.mean(dim=1)

# -------- 2) subspace на train --------
cpca = _fit_cpca_pooled(neg_tr, pos_tr, SUBSPACE_DIM, MIN_EXPLAINED, PCA_DELTA_MODE)
V_full = cpca.U.float()
d_store = V_full.shape[1]
d_eff = min(d_store, MONGE_DIM_CAP)
V = V_full[:, :d_eff].contiguous()   # (C, d)
mu_neg_c = cpca.mu_neg.float()       # (C,)

# -------- 3) z на train / val (для Monge ковариаций — train) --------
z_neg_tr = _z_coords(h_neg_tr, mu_neg_c, V)
z_pos_tr = _z_coords(h_pos_tr, mu_neg_c, V)
z_neg_va = _z_coords(h_neg_va, mu_neg_c, V)
z_pos_va = _z_coords(h_pos_va, mu_neg_c, V)

mu_neg_z = z_neg_tr.mean(dim=0)
mu_pos_z = z_pos_tr.mean(dim=0)

Sigma_neg = _covariance(z_neg_tr)
Sigma_pos = _covariance(z_pos_tr)
M = _monge_matrix(Sigma_neg, Sigma_pos)

# -------- 4) mean-diff: оригинал C и проекция d --------
delta_mean_c_global = h_pos.mean(dim=0) - h_neg.mean(dim=0)           # (C,)
delta_mean_c_train = h_pos_tr.mean(dim=0) - h_neg_tr.mean(dim=0)      # (C,) — согласовано с fit на train
delta_mean_z_global = delta_mean_c_global @ V                        # (d,)
delta_mean_z_train = delta_mean_c_train @ V                            # (d,)

# -------- 5) Monge: сдвиг в z относительно «оставить z» --------
# Попарно на val: T(z)=mu_pos_z + M@(z-mu_neg_z); приращение Δz = T(z)-z
z_monge_va = mu_pos_z + (z_neg_va - mu_neg_z) @ M.T
delta_monge_z_per_sample = z_monge_va - z_neg_va                      # (Nva, d)
delta_monge_z_mean = delta_monge_z_per_sample.mean(dim=0)              # (d,)

# Точка «средний neg» в z (train): куда тянет Monge vs mean-diff сдвиг центроидов
z_bar_neg = z_neg_tr.mean(dim=0)
z_bar_pos = z_pos_tr.mean(dim=0)
z_tilde_bar_neg = mu_pos_z + (z_bar_neg - mu_neg_z) @ M.T
delta_monge_z_at_bar_neg = z_tilde_bar_neg - z_bar_neg                # (d,)

delta_monge_c_mean = _lift_z_to_c(delta_monge_z_mean.unsqueeze(0), V).squeeze(0)           # (C,)
delta_monge_c_at_bar = _lift_z_to_c(delta_monge_z_at_bar_neg.unsqueeze(0), V).squeeze(0) # (C,)

# Для сравнения: mean-diff только как сдвиг центроидов в z (должен совпасть с mu_pos_z - mu_neg_z при больших N)
delta_centroids_z = mu_pos_z - mu_neg_z                               # (d,)

out = {
    "meta": {
        "n": n,
        "n_train": n_train,
        "n_val": n - n_train,
        "C": C,
        "d": d_eff,
        "d_cpca_full": d_store,
        "pca_delta_mode": PCA_DELTA_MODE,
    },
    # оригинал R^C
    "mean_diff_delta_c_global": delta_mean_c_global.detach().cpu(),
    "mean_diff_delta_c_train": delta_mean_c_train.detach().cpu(),
    "monge_delta_c_mean_val": delta_monge_c_mean.detach().cpu(),
    "monge_delta_c_at_zbar_neg": delta_monge_c_at_bar.detach().cpu(),
    # проекция R^d
    "V_d": V.detach().cpu(),
    "mu_neg_c": mu_neg_c.detach().cpu(),
    "mu_neg_z": mu_neg_z.detach().cpu(),
    "mu_pos_z": mu_pos_z.detach().cpu(),
    "M": M.detach().cpu(),
    "mean_diff_delta_z_global": delta_mean_z_global.detach().cpu(),
    "mean_diff_delta_z_train": delta_mean_z_train.detach().cpu(),
    "delta_centroids_z": delta_centroids_z.detach().cpu(),
    "monge_delta_z_per_sample_val": delta_monge_z_per_sample.detach().cpu(),
    "monge_delta_z_mean_val": delta_monge_z_mean.detach().cpu(),
    "monge_delta_z_at_zbar_neg": delta_monge_z_at_bar_neg.detach().cpu(),
    # сырые z/h на val (по желанию смотреть)
    "h_neg_val": h_neg_va.detach().cpu(),
    "h_pos_val": h_pos_va.detach().cpu(),
    "z_neg_val": z_neg_va.detach().cpu(),
    "z_pos_val": z_pos_va.detach().cpu(),
    "z_monge_val": z_monge_va.detach().cpu(),
}

if SAVE_PATH is not None:
    torch.save(out, SAVE_PATH)
    print("saved:", SAVE_PATH.resolve())

print("C =", C, "d =", d_eff)
print("mean_diff_delta_c_global", tuple(out["mean_diff_delta_c_global"].shape))
print("mean_diff_delta_z_global ", tuple(out["mean_diff_delta_z_global"].shape))
print("monge_delta_z_mean_val   ", tuple(out["monge_delta_z_mean_val"].shape))
print("monge_delta_c_mean_val   ", tuple(out["monge_delta_c_mean_val"].shape))
```

```python
import torch
from src.steering.subspace_act import ContrastivePCATokenWise

# =========================
# Config
# =========================
N_SAMPLES = 512
TRAIN_FRAC = 0.8
SUBSPACE_DIM = 32
MIN_EXPLAINED = 0.9
PCA_DELTA_MODE = "raw_delta"
MONGE_DIM_CAP = 16
EPS = 1e-6
BRANCHES = ["pooled", "sequence"]


def _to_ntc(x: torch.Tensor) -> torch.Tensor:
    # (N,C) -> (N,1,C), (N,T,C) -> (N,T,C)
    x = x.float()
    if x.dim() == 2:
        x = x.unsqueeze(1)
    if x.dim() != 3:
        raise ValueError(f"Expected (N,C) or (N,T,C), got {tuple(x.shape)}")
    return x

def _fit_cpca_like_extraction(neg_ntc, pos_ntc, subspace_dim, min_explained, pca_delta_mode):
    finalize_append = pca_delta_mode == "append_mean_dir"
    cpca = ContrastivePCATokenWise(
        subspace_dim=subspace_dim,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_ntc, pos_ntc)
    evr = cpca.explained_variance_ratio()
    if evr.numel() == 0:
        d_sel = 1
    else:
        d_sel = int((evr.cumsum(0) < min_explained).sum().item()) + 1
        d_sel = max(1, min(d_sel, cpca.U.shape[1]))
    cpca = ContrastivePCATokenWise(
        subspace_dim=d_sel,
        pca_delta_mode=pca_delta_mode,
    ).fit(neg_ntc, pos_ntc, finalize_append_mean=finalize_append)
    return cpca

def _fit_neutral_basis_lowd(Z_neutral_ntd, kmax):
    # Z_neutral_ntd: (N,T,d)
    X = Z_neutral_ntd.reshape(-1, Z_neutral_ntd.shape[-1])       # (N*T,d)
    Xc = X - X.mean(dim=0, keepdim=True)
    _, _, Vh = torch.linalg.svd(Xc, full_matrices=False)
    k_eff = min(kmax, Vh.shape[0], Vh.shape[1])
    return Vh[:k_eff].T.contiguous()                              # (d,k_eff)

def _overlap_tokens_vs_k_lowd(delta_tok_d, U_neutral_low, eps=1e-12):
    # delta_tok_d: (T,d), U_neutral_low: (d,K)
    denom = delta_tok_d.norm(dim=1, keepdim=True).clamp(min=eps)
    cols = []
    K = U_neutral_low.shape[1]
    for k in range(1, K + 1):
        Uk = U_neutral_low[:, :k]                                 # (d,k)
        proj = (delta_tok_d @ Uk) @ Uk.T                          # (T,d)
        ov = proj.norm(dim=1, keepdim=True) / denom
        cols.append(ov)
    return torch.cat(cols, dim=1)     

def _safe_mu_neg(mu_neg: torch.Tensor, c_dim: int) -> torch.Tensor:
    # cpca.mu_neg может быть (C,), (1,C) или (T,C)
    mu_neg = mu_neg.float()
    if mu_neg.dim() == 1:
        out = mu_neg
    elif mu_neg.dim() == 2:
        out = mu_neg.mean(dim=0)
    else:
        raise ValueError(f"Unexpected mu_neg shape: {tuple(mu_neg.shape)}")
    if out.numel() != c_dim:
        raise ValueError(f"mu_neg mismatch: got {out.numel()}, expected {c_dim}")
    return out


def compare_branch(pos_x, neg_x, branch_name):
    pos_ntc = _to_ntc(pos_x)
    neg_ntc = _to_ntc(neg_x)

    n = min(N_SAMPLES, pos_ntc.shape[0], neg_ntc.shape[0])
    if n < 6:
        raise ValueError(f"{branch_name}: need at least 6 samples, got {n}")

    pos_ntc = pos_ntc[:n]
    neg_ntc = neg_ntc[:n]

    n_train = max(4, int(n * TRAIN_FRAC))
    n_train = min(n_train, n - 2)

    pos_tr, pos_va = pos_ntc[:n_train], pos_ntc[n_train:n]
    neg_tr, neg_va = neg_ntc[:n_train], neg_ntc[n_train:n]

    # твоя функция fit из ноутбука
    cpca = _fit_cpca_like_extraction(
        neg_tr,
        pos_tr,
        subspace_dim=SUBSPACE_DIM,
        min_explained=MIN_EXPLAINED,
        pca_delta_mode=PCA_DELTA_MODE,
    )

    V = cpca.U.float()
    if V.shape[1] == 0:
        raise ValueError(f"{branch_name}: CPCA returned empty subspace")
    d_eff = min(V.shape[1], MONGE_DIM_CAP)
    V = V[:, :d_eff].contiguous()

    C = pos_tr.shape[-1]
    mu_neg_c = _safe_mu_neg(cpca.mu_neg, C)

    # token pooling -> (N,C)
    h_neg_tr = neg_tr.mean(dim=1)
    h_pos_tr = pos_tr.mean(dim=1)
    h_neg_va = neg_va.mean(dim=1)
    h_pos_va = pos_va.mean(dim=1)

    # z-space
    z_neg_tr = (h_neg_tr - mu_neg_c) @ V
    z_pos_tr = (h_pos_tr - mu_neg_c) @ V
    z_neg_va = (h_neg_va - mu_neg_c) @ V
    z_pos_va = (h_pos_va - mu_neg_c) @ V

    mu_neg_z = z_neg_tr.mean(dim=0)
    mu_pos_z = z_pos_tr.mean(dim=0)

    Sigma_neg = _covariance(z_neg_tr, eps=EPS)
    Sigma_pos = _covariance(z_pos_tr, eps=EPS)
    M = _monge_matrix(Sigma_neg, Sigma_pos, eps=EPS)

    # mean-diff baseline
    delta_mu = mu_pos_z - mu_neg_z
    z_pred_mean = z_neg_va + delta_mu

    # Monge map
    z_pred_monge = mu_pos_z + (z_neg_va - mu_neg_z) @ M.T

    # metric
    err_mean = (z_pred_mean - z_pos_va).norm(dim=1).mean().item()
    err_monge = (z_pred_monge - z_pos_va).norm(dim=1).mean().item()
    gain = err_mean - err_monge  # >0 => Monge better

    print(
        f"{branch_name:8s} | d={d_eff:2d} | Ntr={n_train:3d} Nva={n-n_train:3d} | "
        f"mean-diff={err_mean:.4f} | monge={err_monge:.4f} | gain={gain:+.4f}"
    )

    return {
        "branch": branch_name,
        "d": d_eff,
        "n_train": n_train,
        "n_val": n - n_train,
        "cv_mean_diff": err_mean,
        "cv_monge": err_monge,
        "cv_gain": gain,
        "M": M.detach().cpu(),
        "mu_neg_z": mu_neg_z.detach().cpu(),
        "mu_pos_z": mu_pos_z.detach().cpu(),
        "V_d": V.detach().cpu(),
        "mu_neg_c": mu_neg_c.detach().cpu(),
    }


# =========================
# Run
# =========================
results = {}
for b in BRANCHES:
    results[b] = compare_branch(pos_data[b], neg_data[b], b)

print("\n=== Summary ===")
for b in BRANCHES:
    r = results[b]
    print(f"{b:8s}: mean-diff={r['cv_mean_diff']:.4f}, monge={r['cv_monge']:.4f}, gain={r['cv_gain']:+.4f}")
```
